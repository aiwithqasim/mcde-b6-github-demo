## Q1. Accessing List Elements

```python
numbers = [10, 20, 30, 40, 50, 60, 70]


print(numbers[:: 3 ])
```

##Q2. List Slicing


```python
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]

first_five = numbers[: 5]
last_four = numbers[-4:]
every_second = numbers[:: 2]

print("first five numbers",first_five)
print("last four numbers",last_four)
print( "every second number",every_second)
```

## Q3. List Modification


```python
fruits = ["apple", "banana", "cherry", "mango"]

fruits[1] = "orange"
fruits.insert(2, "grapes")
fruits.remove("cherry")
print(fruits)
```

## Q4. Reverse Without reverse()


```python
numbers = [12, 45, 23, 67, 89, 34]

print("Original list" , numbers)


rotated_list = numbers [ ::-1]
print("rotated list", rotated_list )
```

## Q5. Second Largest Unique Number


```python
numbers = [45, 12, 89, 34, 89, 67, 45, 23]

unique_num = list(set(numbers))

unique_num.sort()

second_largest = unique_num [-2]
print(second_largest)
```

## Q6. Rotate a List


```python
numbers = [10, 20, 30, 40, 50, 60]
k = 2

rotated_list = numbers[ -k: ]  + numbers [ : -k]

print( rotated_list)
```

## Q7. Remove Duplicates, Preserve Order


```python
numbers = [10, 20, 10, 30, 20, 40, 30, 50]

rem_duplicates = list(dict.fromkeys(numbers))

print(rem_duplicates)
```

## Q8. Flatten a Nested List


```python
data = [[10, 20], [30, 40], [50, 60]]

flat_list = sum (data, [])

print(flat_list)
```

## Q10. List Transformation Challenge


```python
data = [15, 5, 25, 10, 35, 20, 30]

unique_number = list(set(data))

unique_number.sort()

three_largest = unique_number[-3:]
first_twos = unique_number[:2]

print(three_largest + first_twos)
```
