# Q22. Set Operations


```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

union = a.union(b)
intersection = a.intersection(b)
difference = a.difference(b)
sym_diff = a.symmetric_difference(b)

print("Union:", union)
print("Intersection:", intersection)
print("Difference (a - b):", difference)
print("Symmetric Difference:", sym_diff)
```

# Q24. Common Students


```python
python_students = {"Ali", "Sara", "Ahmed", "Zain"}
sql_students = {"Sara", "Ahmed", "Hamza", "Usman"}

# Students who took both courses
both_courses_std = python_students.intersection(sql_students)

# Students who took exactly one course
exactly_one_course_std = python_students.symmetric_difference(sql_students)

print("Students who took both Python and SQL:", both_courses_std)
print("Students who took exactly one course:", exactly_one_course_std)
```

# Q25. Subset and Superset


```python
required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}

is_subset = required_skills.issubset(candidate_skills)
is_superset = required_skills.issuperset(candidate_skills)

print("Is required_skills a subset of candidate_skills? \n",is_subset)
print("Is required_skills a superset of candidate_skills? \n",is_superset)
```

# Q27. Three-Set Analysis


```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

#Students in all three
all_three_std = python.intersection(sql).intersection(aws)

#Exactly Python and SQL (not AWS)
python_sql_not_aws = python.intersection(sql).difference(aws)

#Students in at least two
pair_py_sql = python.intersection(sql)
pair_py_aws = python.intersection(aws)
pair_sql_aws = sql.intersection(aws)
at_least_two = pair_py_sql.union(pair_py_aws).union(pair_sql_aws)

print("Students in all three:", all_three_std)
print("Students in exactly Python and SQL (not AWS):", python_sql_not_aws)
print("Students in at least two courses:", at_least_two)
```

# Q28. Transaction Analysis


```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

#Customers who appear in all three days
all_days = day1.intersection(day2).intersection(day3)

pair_1_2 = day1.intersection(day2)
pair_2_3 = day2.intersection(day3)
pair_1_3 = day1.intersection(day3)

#Customers who appear in at least two days
at_least_two = pair_1_2.union(pair_2_3).union(pair_1_3)

#Customers who appear in exactly one day
total_union = day1.union(day2).union(day3)
only_one_day = total_union.difference(at_least_two)

print("Customers who appear in all three days:", all_days)
print("Customers who appear in exactly one day:", only_one_day)
print("Customers who appear in at least two days:", at_least_two)
```

# Q29. Access Control


```python
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing_permission = required_permissions.difference(user_permissions)
extra_permission = user_permissions.difference(required_permissions)
full_authorization = required_permissions.issubset(user_permissions)

print("Missing permissions:", missing_permission )
print("Extra permissions:", extra_permission)
print("authorization:", full_authorization)
```

# Q30. Customer Segmentation


```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

#Customers on all three platforms
all_three = website.intersection(mobile).intersection(store)

# Intersections between pairs
pair_web_mob = website.intersection(mobile)
pair_mob_store = mobile.intersection(store)
pair_web_store = website.intersection(store)

# Customers on at least two platforms
at_least_two = pair_web_mob.union(pair_mob_store).union(pair_web_store)

#Customers on exactly two platforms (at least two excluding all three)
exactly_two = at_least_two.difference(all_three)

#Customers on exactly one platform (total union excluding at least two)
total_union = website.union(mobile).union(store)
exactly_one = total_union.difference(at_least_two)

print("Customers on exactly one platform:", exactly_one)
print("Customers on exactly two platforms:", exactly_two)
print("Customers on all three platforms:", all_three)
```
