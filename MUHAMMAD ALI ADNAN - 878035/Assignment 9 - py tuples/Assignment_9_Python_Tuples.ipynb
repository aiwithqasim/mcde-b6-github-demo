# Q17. Nested Tuple Access


```python
students = (
 ("Ali", (85, 90, 88)),
 ("Sara", (92, 87, 95)),
 ("Ahmed", (78, 84, 80))
)

sara_grades = students[1][1][1]
ali_grade_avg = sum(students[0][1]) // len(students[0][1])
ahmed_sorted_grade = list(students[2][1])
ahmed_sorted_grade.sort()
ahmed_high_grade = ahmed_sorted_grade[-1]


print("sara gardes :", sara_grades)
print("avg of ali :", ali_grade_avg)
print("highest grade of ahmed :", ahmed_high_grade)
```

# Q18. Tuple Sorting



```python
employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

# Helper function to get the salary (index 1) without lambdas or imports
def get_salary(employee_item):
    return employee_item[1]

sorted_employees = tuple(sorted(employees, key=get_salary, reverse=True))
print(sorted_employees)
```

# Q19. Tuple Matrix


```python
matrix = (
    (2, 4, 6),
    (8, 10, 12),
    (14, 16, 18)
)

diagonal= [matrix[0][0] , matrix[1][1], matrix[2][2]]
diagonal_product = matrix[0][0] * matrix[1][1] * matrix[2][2]
row_sum = [sum(matrix[0]) , sum(matrix[1]) , sum(matrix[2])]


print("diagonal :", diagonal)
print("diagonal product :", diagonal_product)
print("row sum :", row_sum)
```

# Q20. Inventory Tuple Challenge


```python
records = (
 ("Laptop", 120000, 5),
 ("Mouse", 1500, 20),
 ("Keyboard", 3500, 10)
)

# Unpack each item individually without loops
item1, item2, item3 = records

# 1. Calculate total inventory value
val1 = item1[1] * item1[2]
val2 = item2[1] * item2[2]
val3 = item3[1] * item3[2]
total_inventory_value = val1 + val2 + val3

# 2. Find the most expensive product (using max with a key function without loops)
def get_price(item):
    return item[1]

most_expensive_product = max(item1, item2, item3, key=get_price)[0]

# 3. Build reversed tuple of names
product_names = (item1[0], item2[0], item3[0])
reversed_names = product_names[::-1]

print("Total Inventory Value:", total_inventory_value)
print("Most Expensive Product:", most_expensive_product)
print("Reversed Product Names:", reversed_names)
```
