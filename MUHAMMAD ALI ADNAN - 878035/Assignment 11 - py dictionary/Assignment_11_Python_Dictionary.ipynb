# Q35. Student Marks


```python
marks = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zain": 88
}

highest_marks = max(marks.values())
lowest_marks = min(marks.values())
avg_marks= sum(marks.values())  // len(marks.values())
top_scorring = max(marks, key=marks.get)

print("Top Scorring Student :", top_scorring)
print("Highest Marks :", highest_marks)
print("Lowest Marks :", lowest_marks)
print("Average Marks :", avg_marks)


```

# Q38. Order Calculation


```python
order = {
 "laptop": {"price": 120000, "quantity": 2},
 "mouse": {"price": 1500, "quantity": 5},
 "keyboard": {"price": 3500, "quantity": 3}
}
laptop_subtotal = order["laptop"]["price"] * order["laptop"]["quantity"]
mouse_subtotal = order["mouse"]["price"] * order["mouse"]["quantity"]
keyboard_subtotal = order["keyboard"]["price"] * order["keyboard"]["quantity"]
grand_total = laptop_subtotal + mouse_subtotal + keyboard_subtotal
after_disc = int(grand_total * 0.9)

print("laptop subtotals :" , laptop_subtotal)
print("mouse subtotals :" , mouse_subtotal)
print("laptop subtotals :" , laptop_subtotal)
print("grand total :" , grand_total)
print("after discount :" , after_disc)
```

# Q39. Employee Performance


```python
employees = {
 "Ali": {"python": 90, "sql": 85, "aws": 88},
 "Sara": {"python": 95, "sql": 92, "aws": 90},
 "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

# Averages
ali_avg_marks=  sum(employees['Ali'].values()) //len(employees["Ali"].values())
sara_avg_marks=  sum(employees['Sara'].values()) //len(employees["Sara"].values())
ahmed_avg_marks=  sum(employees['Ahmed'].values()) //len(employees["Ahmed"].values())

# Helper function to get average marks
def get_avg_marks(employee_name):
    return sum(employees[employee_name].values()) / len(employees[employee_name])

# Top performer
top_performer = max(employees, key=get_avg_marks)

# hisghest Score
highest_sql_score = max(
    employees["Ali"]["sql"],
    employees["Sara"]["sql"],
    employees["Ahmed"]["sql"]
)

print("Highest SQL Score:", highest_sql_score)
print("Top Performer :", top_performer)
print("Ali's Avg marks :", ali_avg_marks)
print("sara's Avg marks :", sara_avg_marks)
print("ahmed's Avg marks :", ahmed_avg_marks)
```

# Q40. Inventory Management


```python
inventory = {
 "P101": {"name": "Laptop", "price": 120000, "stock": 5},
 "P102": {"name": "Mouse", "price": 1500, "stock": 20},
 "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

# Update stock after selling two laptops and five mice
inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5

# Calculate total value of the inventory
total_value = sum(item["price"] * item["stock"] for item in inventory.values())

# Helper function to get the stock value of an item key
def get_item_value(item_id):
    return inventory[item_id]["price"] * inventory[item_id]["stock"]

# Find the most valuable item by price x stock without lambda
most_valuable_key = max(inventory, key=get_item_value)
most_valuable_item = inventory[most_valuable_key]

print("Updated Inventory:", inventory)
print("Total Inventory Value:", total_value)
print("Most Valuable Item:", most_valuable_item["name"], "(Value:", most_valuable_item["price"] * most_valuable_item["stock"], ")")
```
