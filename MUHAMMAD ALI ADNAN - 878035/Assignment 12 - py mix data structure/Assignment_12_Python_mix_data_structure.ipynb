# Q41. Student Performance System



```python
students = [
 {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
 {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
 {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

# Averges marks
avg_id1= sum(students[0]["marks"]) // len(students[0]["marks"])
avg_id2= sum(students[1]["marks"]) // len(students[1]["marks"])
avg_id3= sum(students[2]["marks"]) // len(students[2]["marks"])

# top Student
total = max(
    students[0]["marks"],
    students[1]["marks"],
    students[2]["marks"],
)
top_std = max(total)
pos= total.index(top_std)
name= students[pos]["name"]

# Tuple of Names
tuple_of_names = tuple([students[0]["name"] , students[1]["name"] , students[2]["name"]])

# Set of Unique marks
set_of_marks = set(students[0]["marks"]+ students[1]["marks"] + students[2]["marks"])


print("Average marks od id 1 :", avg_id1)
print("Average marks od id 2 :", avg_id2)
print("Average marks od id 3 :", avg_id3)
print("Top Student is  :", name)
print("Tuples of name :", tuple_of_names )
print("Set of unique marks :", set_of_marks )
```

# Q42. Employee Skills


```python
employees = {
 "Ali": {"Python", "SQL", "AWS"},
 "Sara": {"Python", "Databricks", "SQL"},
 "Ahmed": {"AWS", "Snowflake", "SQL"}
}

# Common Skils
ali_skill = employees["Ali"]
sara_skill = employees["Sara"]
ahmed_skill = employees["Ahmed"]
common_skill = ali_skill.intersection(sara_skill).intersection(ahmed_skill)

# only ali's skill
ali_skill = employees["Ali"]

# All teams Skill
all_teams_skill = ali_skill.union(sara_skill).union(ahmed_skill)

# python & sql
required = {"Python", "SQL"}

names = list(employees.keys())
skill_sets = list(employees.values())

qualified = []
if required <= skill_sets[0]: qualified.append(names[0])
if required <= skill_sets[1]: qualified.append(names[1])
if required <= skill_sets[2]: qualified.append(names[2])


print("Common skill is :", common_skill)
print("Only Ali's Skill :" , ali_skill)
print("All teams Skills :" , all_teams_skill)
print("Python + SQL Skill Qualified :", qualified)
print("Count:", len(qualified))
```

# Q43. Sales Data


```python
sales = [
    ("Laptop",   120000, 2),
    ("Mouse",    1500,   5),
    ("Keyboard", 3500,   3)
]

names = [sales[0][0], sales[1][0], sales[2][0]]

revenues = [
    sales[0][1] * sales[0][2],
    sales[1][1] * sales[1][2],
    sales[2][1] * sales[2][2]
]

# 1. Total revenue
total_revenue = sum(revenues)

# 2. Top-revenue product
top_product = names[revenues.index(max(revenues))]

# 3. Name-to-revenue dictionary
name_to_revenue = dict(zip(names, revenues))

# 4. Set of prices
prices = {sales[0][1], sales[1][1], sales[2][1]}

print("Total revenue:", total_revenue)
print("Top product:", top_product)
print("Name to revenue:", name_to_revenue)
print("Prices:", prices)
```

# Q44. Customer Orders


```python
customers = {
    "C101": {"name": "Ali",   "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara",  "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}

ids = sorted(customers.keys())

names = [
    customers[ids[0]]["name"],
    customers[ids[1]]["name"],
    customers[ids[2]]["name"]
]

totals = [
    sum(customers[ids[0]]["orders"]),
    sum(customers[ids[1]]["orders"]),
    sum(customers[ids[2]]["orders"])
]

spending = dict(zip(names, totals))
top_customer = names[totals.index(max(totals))]
totals_tuple = tuple(totals)

print("Spending:", spending)
print("Top customer:", top_customer)
print("Totals tuple:", totals_tuple)
```

#Q45. Course Enrollment


```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL":    {"Sara", "Ahmed", "Zain"},
    "AWS":    {"Ali", "Ahmed", "Zain"}
}

sets = list(courses.values())
p, s, a = sets[0], sets[1], sets[2]

all_three   = p & s & a
in_two_plus = (p & s) | (p & a) | (s & a)
exactly_two = in_two_plus - all_three
exactly_one = (p | s | a) - in_two_plus
unique_count = len(p | s | a)

print("Exactly two:", sorted(exactly_two))
print("All three:", sorted(all_three))
print("Exactly one:", sorted(exactly_one))
print("Unique students:", unique_count)
```

# Q46. E-Commerce Transactions


```python
transactions = [
 {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
 {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
 {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]


ids   = [transactions[0]["id"], transactions[1]["id"], transactions[2]["id"]]
custs = [transactions[0]["customer"], transactions[1]["customer"], transactions[2]["customer"]]
i0, i1, i2 = transactions[0]["items"], transactions[1]["items"], transactions[2]["items"]

totals = [
    i0[0][1]*i0[0][2] + i0[1][1]*i0[1][2],
    i1[0][1]*i1[0][2] + i1[1][1]*i1[1][2],
    i2[0][1]*i2[0][2]
]

total_revenue = sum(totals)
ali_spending  = (custs[0] == "Ali")*totals[0] + (custs[1] == "Ali")*totals[1] + (custs[2] == "Ali")*totals[2]
largest_id    = ids[totals.index(max(totals))]
products      = {i0[0][0], i0[1][0], i1[0][0], i1[1][0], i2[0][0]}

print("Total revenue:", total_revenue)
print("Ali spending:", ali_spending)
print("Largest transaction:", largest_id, max(totals))
print("Unique products:", sorted(products))
```

# Q47. Pipeline Monitoring

```python
pipelines = {
 "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
 "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
 "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}
names = list(pipelines.keys())
vals  = list(pipelines.values())
recs  = [vals[0]["records"], vals[1]["records"], vals[2]["records"]]
totals = [sum(recs[0]), sum(recs[1]), sum(recs[2])]

failed = []
if vals[0]["status"] == "failed": failed.append(names[0])
if vals[1]["status"] == "failed": failed.append(names[1])
if vals[2]["status"] == "failed": failed.append(names[2])

averages = dict(zip(names, [
    round(totals[0] / len(recs[0]), 2),
    round(totals[1] / len(recs[1]), 2),
    round(totals[2] / len(recs[2]), 2)
]))

print("Total processed:", sum(totals))
print("Failed:", failed)
print("Largest:", names[totals.index(max(totals))], max(totals))
print("Average per run:", averages)
```

# Q48. Banking Transactions

```python
accounts = {
 "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
 "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
 "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

vals    = list(accounts.values())
holders = [vals[0]["holder"], vals[1]["holder"], vals[2]["holder"]]
tx      = [vals[0]["transactions"], vals[1]["transactions"], vals[2]["transactions"]]

balances = [sum(tx[0]), sum(tx[1]), sum(tx[2])]
final    = dict(zip(holders, balances))

all_tx = tx[0] + tx[1] + tx[2]          # list concatenation, 10 items
withdrawals = abs(
    min(all_tx[0], 0) + min(all_tx[1], 0) + min(all_tx[2], 0) + min(all_tx[3], 0) +
    min(all_tx[4], 0) + min(all_tx[5], 0) + min(all_tx[6], 0) + min(all_tx[7], 0) +
    min(all_tx[8], 0) + min(all_tx[9], 0)
)

print("Final balances:", final)
print("Highest balance:", holders[balances.index(max(balances))], max(balances))
print("Total withdrawals:", withdrawals)
print("Largest deposit:", max(all_tx))
```

# Q49. Product Recommendations

```python
users = {
 "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
 "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
 "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}
vals = list(users.values())
pur  = [vals[0]["purchased"], vals[1]["purchased"], vals[2]["purchased"]]
wish = [vals[0]["wishlist"], vals[1]["wishlist"], vals[2]["wishlist"]]

bought_2plus  = (pur[0] & pur[1]) | (pur[0] & pur[2]) | (pur[1] & pur[2])
ali_wish_only = wish[0] - pur[0]
shared_wish   = (wish[0] & wish[1]) | (wish[0] & wish[2]) | (wish[1] & wish[2])
all_wish3     = wish[0] & wish[1] & wish[2]
all_products  = pur[0] | pur[1] | pur[2] | wish[0] | wish[1] | wish[2]

print("Bought by 2+ users:", sorted(bought_2plus))
print("Ali wishlist minus purchases:", sorted(ali_wish_only))
print("Shared wishlist (2+ users):", sorted(shared_wish))
print("In all 3 wishlists:", sorted(all_wish3))
print("All products:", sorted(all_products))
```

# Q50. Final Company Analysis

```python
company = {
 "departments": {
 "Engineering": {
 "employees": [
 ("Ali", 120000, {"Python", "AWS", "SQL"}),
 ("Sara", 150000, {"Python", "Databricks", "SQL"})
 ],
 "budget": 500000
 },
 "Analytics": {
 "employees": [
 ("Ahmed", 100000, {"SQL", "Power BI"}),
 ("Zain", 130000, {"Python", "SQL", "Tableau"})
 ],
 "budget": 400000
 }
 }
}


depts  = company["departments"]
dnames = list(depts.keys())
dvals  = list(depts.values())

e0, e1 = dvals[0]["employees"], dvals[1]["employees"]
budgets = [dvals[0]["budget"], dvals[1]["budget"]]
allp = e0 + e1                                    # 4 employee tuples

names    = [allp[0][0], allp[1][0], allp[2][0], allp[3][0]]
salaries = [allp[0][1], allp[1][1], allp[2][1], allp[3][1]]
skills   = [allp[0][2], allp[1][2], allp[2][2], allp[3][2]]

payroll = [e0[0][1] + e0[1][1], e1[0][1] + e1[1][1]]          # 1
company_payroll = sum(payroll)                                 # 2
top_paid = names[salaries.index(max(salaries))]                # 3
shared_all = skills[0] & skills[1] & skills[2] & skills[3]     # 4
eng_only = (skills[0] | skills[1]) - (skills[2] | skills[3])   # 5
remaining = dict(zip(dnames, [budgets[0] - payroll[0], budgets[1] - payroll[1]]))  # 6
avgs = [payroll[0] / len(e0), payroll[1] / len(e1)]
higher_avg = dnames[avgs.index(max(avgs))]                     # 7
name_to_salary = dict(zip(names, salaries))                    # 8
all_skills = skills[0] | skills[1] | skills[2] | skills[3]     # 9
sufficient = sum(budgets) >= company_payroll                   # 10

print("1. Payroll per dept:", dict(zip(dnames, payroll)))
print("2. Company payroll:", company_payroll)
print("3. Highest paid:", top_paid, max(salaries))
print("4. Skills shared by all:", sorted(shared_all))
print("5. Engineering-only skills:", sorted(eng_only))
print("6. Remaining budget:", remaining)
print("7. Higher avg salary:", higher_avg, max(avgs))
print("8. Name to salary:", name_to_salary)
print("9. All unique skills:", sorted(all_skills))
print("10. Budget sufficient:", sufficient, "| Combined budget:", sum(budgets))
```
