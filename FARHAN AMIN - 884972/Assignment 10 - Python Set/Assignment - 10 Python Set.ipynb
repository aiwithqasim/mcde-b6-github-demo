### Q21. Unique Values
Convert to a set and count unique elements.

```python
numbers = [10, 20, 10, 30, 40, 20, 50]

unique_numbers = set(numbers)
unique_count = len(unique_numbers)

print("Unique elements:", unique_numbers)
print("Count:", unique_count)
```

### Q22. Set Operations
Find union, intersection, a - b, and symmetric difference.

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

set_union = a | b
set_intersection = a & b
set_difference = a - b
set_symmetric_difference = a ^ b

print("Union:", set_union)
print("Intersection:", set_intersection)
print("Difference (a - b):", set_difference)
print("Symmetric Difference:", set_symmetric_difference)
```

### Q24. Common Students
Find students taking both and exactly one course.

```python
python_students = {"Ali", "Sara", "Ahmed", "Zain"}
sql_students = {"Sara", "Ahmed", "Hamza", "Usman"}

both_courses = python_students & sql_students
exactly_one_course = python_students ^ sql_students

print("Students in both courses:", both_courses)
print("Students in exactly one course:", exactly_one_course)
```

### Q25. Subset and Superset
Check required skills are a subset and candidate skills a superset.

```python
required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}

if required_skills.issubset(candidate_skills):
    print(f"candidates have required skills {required_skills} ")
if candidate_skills.issuperset(required_skills):
    extra_skills = candidate_skills - required_skills
    print(f"candidates have {extra_skills} extra skills")
```

### Q26. Missing Numbers
Find missing integers from 1 through 10 using set operations.

```python
numbers = {1, 2, 4, 6, 8, 10}

full_range = set(range(1, 11))
missing_numbers = full_range - numbers

print("Missing numbers:", missing_numbers)
```

### Q27. Three-Set Analysis
Find students in all three, exactly Python and SQL (not AWS), and at least two.

```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

all_three = python & sql & aws
python_sql_not_aws = (python & sql) - aws
at_least_two = (python & sql) | (python & aws) | (sql & aws)

print("Students in all three:", all_three)
print("Students in exactly Python and SQL (not AWS):", python_sql_not_aws)
print("Students in at least two:", at_least_two)
```

### Q28. Transaction Analysis
Find IDs appearing on all days, exactly one day, and at least two days.

```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

all_days = day1 & day2 & day3
at_least_two = (day1 & day2) | (day2 & day3) | (day1 & day3)
all_elements = day1 | day2 | day3
exactly_one = all_elements - at_least_two

print("Appearing on all days:", all_days)
print("Appearing on exactly one day:", exactly_one)
print("Appearing on at least two days:", at_least_two)
```

### Q29. Access Control
Find missing and extra permissions and test full authorization.

```python
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing = required_permissions - user_permissions
extra = user_permissions - required_permissions
is_authorized = required_permissions.issubset(user_permissions)

print("Missing permissions:", missing)
print("Extra permissions:", extra)
print("Is authorized:", is_authorized)
```

### Q30. Customer Segmentation
Find customers on exactly one, exactly two, and all three platforms.

```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website & mobile & store
at_least_two = (website & mobile) | (mobile & store) | (website & store)
exactly_two = at_least_two - all_three
all_platforms = website | mobile | store
exactly_one = all_platforms - at_least_two

print("Customers on all three:", all_three)
print("Customers on exactly two:", exactly_two)
print("Customers on exactly one:", exactly_one)
```
