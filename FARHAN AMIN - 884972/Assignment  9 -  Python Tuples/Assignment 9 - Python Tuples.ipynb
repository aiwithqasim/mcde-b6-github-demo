```python
# ###Q20. Inventory Tuple Challenge
# records = (
#     ("Laptop", 120000, 5),
#     ("Mouse", 1500, 20),
#     ("Keyboard", 3500, 10)
# )
# Calculate inventory value, find most expensive product, and build reversed tuple of names.

records = (
     ("Laptop", 120000, 5),
     ("Mouse", 1500, 20),
     ("Keyboard", 3500, 10)
)
total_inventory_value = (
        records[0][1] * records[0][2] + records[1][1] * records[1][2] + records[2][1] * records[2][2]
      
        
)

inventory_name_reversed = list((
    records[2][0], records[1][0], records[0][0]
))

most_expensive = records [0][0] 
print(f"most_expensive_product: {most_expensive}")
print(inventory_name_reversed)
print(f"total_inventory_value: {total_inventory_value}")
```

```python
# Q19. Tuple Matrix
# matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))
# Extract diagonal, calculate its product, and build a tuple of row sums

matrix = ((2, 4, 6), 
          (8, 10, 12), 
          (14, 16, 18))

diagonal = matrix[0][0], matrix[1][1], matrix[2][2]
diagonal_product = matrix[0][0] * matrix[1][1] * matrix[2][2]
row_sum = tuple (
                (matrix[0][0]+matrix[0][1]+matrix[0][2], 
                matrix[1][0]+matrix[1][1]+matrix[1][2],
                matrix[2][0]+matrix[2][1]+matrix[2][2])
) 

print(f"diagonal from matrix: {diagonal}")
print(f"diagonal from matrix: {diagonal_product}")
print(f"tuple of row_sums: {row_sum}")
    
```

```python
# Q17. Nested Tuple Access

# students = (
#     ("Ali", (85, 90, 88)),
#     ("Sara", (92, 87, 95)),
#     ("Ahmed", (78, 84, 80))
# )
# Extract Sara second score, Ali average, and Ahmed highest score

students = (
 ("Ali", (85, 90, 88)),
 ("Sara", (92, 87, 95)),
 ("Ahmed", (78, 84, 80))
)

sara_second = students[1][1][1]

ali_avg = sum(students[0][1]) / len(students[0][1])

ahmed_highest = max(students[2][1])

print("Sara's Second Score:", sara_second)
print("Ali's Average:", ali_avg)
print("Ahmed's Highest Score:", ahmed_highest)
```

```python
# Q16. Modify an Immutable Tuple
# numbers = (10, 20, 30, 40, 50)
# Create a new tuple replacing 30 with 300.

numbers = (10, 20, 30, 40, 50)

new_numbers = numbers[:2] + (300,) + numbers[3:]

print("New Tuple:", new_numbers)
```

```python
# Q15. Extended Tuple Unpacking
# data = (100, 200, 300, 400, 500, 600)
# Unpack first two and last into variables; put all remaining elements into a list.

data = (100, 200, 300, 400, 500, 600)

first, second, *remaining, last = data

print("First:", first)
print("Second:", second)
print("Remaining:", remaining)
print("Last:", last)
```

```python
# Q14. Count and Locate
# numbers = (10, 20, 30, 20, 40, 20, 50)
# Count occurrences of 20 and find the first index.

numbers = (10, 20, 30, 20, 40, 20, 50)


count_20 = numbers.count(20)

first_index = numbers.index(20)

print("Count of 20:", count_20)
print("First index of 20:", first_index)

```

```python
# Q13. Tuple Concatenation
# tuple1 = (10, 20, 30)
# tuple2 = (40, 50, 60)
# Combine tuples, then create another tuple in reverse order.

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

combined_tuples = tuple1 + tuple2

reversed_tuple = combined_tuples[::-1]

print("Combined:", combined_tuples)
print("Reversed:", reversed_tuple)
```
