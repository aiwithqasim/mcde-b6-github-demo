## Q41. Student Performance System

```
students = [
 {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
 {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
 {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

# 1. Calculate each average
avg1 = sum(students[0]["marks"]) / len(students[0]["marks"])
avg2 = sum(students[1]["marks"]) / len(students[1]["marks"])
avg3 = sum(students[2]["marks"]) / len(students[2]["marks"])
averages = {students[0]["name"]: round(avg1, 2), students[1]["name"]: round(avg2, 2), students[2]["name"]: round(avg3, 2)}

# 2. Top student
top_student = max((avg1, students[0]["name"]), (avg2, students[1]["name"]), (avg3, students[2]["name"]))[1]

# 3. Tuple of names
names_tuple = (students[0]["name"], students[1]["name"], students[2]["name"])

# 4. Set of unique marks
unique_marks = set(students[0]["marks"]) | set(students[1]["marks"]) | set(students[2]["marks"])

print("Averages:", averages)
print("Top Student:", top_student)
print("Names Tuple:", names_tuple)
print("Unique Marks:", unique_marks)
```

## Q42. Employee Skills

```
employees = {
 "Ali": {"Python", "SQL", "AWS"},
 "Sara": {"Python", "Databricks", "SQL"},
 "Ahmed": {"AWS", "Snowflake", "SQL"}
}

# 1. Common skills
common_skills = employees["Ali"] & employees["Sara"] & employees["Ahmed"]

# 2. Ali-only skills
ali_only = employees["Ali"] - (employees["Sara"] | employees["Ahmed"])

# 3. All team skills
all_team_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]

# 4. Employees meeting Python + SQL requirements
req = {"Python", "SQL"}
python_sql_req = []
req.issubset(employees["Ali"]) and python_sql_req.append("Ali")
req.issubset(employees["Sara"]) and python_sql_req.append("Sara")
req.issubset(employees["Ahmed"]) and python_sql_req.append("Ahmed")

print("Common Skills:", common_skills)
print("Ali-only Skills:", ali_only)
print("All Team Skills:", all_team_skills)
print("Python + SQL Employees:", python_sql_req)
```

## Q43. Sales Data

```
sales = [
 ("Laptop", 120000, 2),
 ("Mouse", 1500, 5),
 ("Keyboard", 3500, 3)
]

# 1. Calculate revenue
rev1 = sales[0][1] * sales[0][2]
rev2 = sales[1][1] * sales[1][2]
rev3 = sales[2][1] * sales[2][2]
total_revenue = rev1 + rev2 + rev3

# 2. Top-revenue product
top_product = max((rev1, sales[0][0]), (rev2, sales[1][0]), (rev3, sales[2][0]))[1]

# 3. Name-to-revenue dictionary
name_to_revenue = {sales[0][0]: rev1, sales[1][0]: rev2, sales[2][0]: rev3}

# 4. Set of prices
set_of_prices = {sales[0][1], sales[1][1], sales[2][1]}

print("Total Revenue:", total_revenue)
print("Top Revenue Product:", top_product)
print("Name-to-Revenue Dict:", name_to_revenue)
print("Set of Prices:", set_of_prices)
```

## Q44. Customer Orders

## Q45. Course Enrollment

```
courses = {
 "Python": {"Ali", "Sara", "Ahmed"},
 "SQL": {"Sara", "Ahmed", "Zain"},
 "AWS": {"Ali", "Ahmed", "Zain"}
}

p = courses["Python"]
s = courses["SQL"]
a = courses["AWS"]

# 1. Students in exactly two courses
exactly_two = (p & s - a) | (s & a - p) | (a & p - s)

# 2. Students in all three courses
all_three = p & s & a

# 3. Students in exactly one course
exactly_one = (p - s - a) | (s - p - a) | (a - p - s)

# 4. Count unique students
unique_students_count = len(p | s | a)

print("Exactly Two Courses:", exactly_two)
print("All Three Courses:", all_three)
print("Exactly One Course:", exactly_one)
print("Unique Students Count:", unique_students_count)
```

## Q46. E-Commerce Transactions

```
transactions = [
 {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
 {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
 {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

# 1. Calculate total revenue
t1_rev = (transactions[0]["items"][0][1] * transactions[0]["items"][0][2]) + (transactions[0]["items"][1][1] * transactions[0]["items"][1][2])
t2_rev = (transactions[1]["items"][0][1] * transactions[1]["items"][0][2]) + (transactions[1]["items"][1][1] * transactions[1]["items"][1][2])
t3_rev = (transactions[2]["items"][0][1] * transactions[2]["items"][0][2])
total_revenue = t1_rev + t2_rev + t3_rev

# 2. Ali spending
ali_spending = t1_rev + t3_rev

# 3. Largest transaction
largest_transaction = max(t1_rev, t2_rev, t3_rev)

# 4. Unique product names
unique_products = {
    transactions[0]["items"][0][0], transactions[0]["items"][1][0],
    transactions[1]["items"][0][0], transactions[1]["items"][1][0],
    transactions[2]["items"][0][0]
}

print("Total Revenue:", total_revenue)
print("Ali Spending:", ali_spending)
print("Largest Transaction:", largest_transaction)
print("Unique Products:", unique_products)
```

## Q47. Pipeline Monitoring

```
pipelines = {
 "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
 "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
 "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

# 1. Total processed records
tot1 = sum(pipelines["pipeline_1"]["records"])
tot2 = sum(pipelines["pipeline_2"]["records"])
tot3 = sum(pipelines["pipeline_3"]["records"])
total_processed_records = tot1 + tot2 + tot3

# 2. Failed pipelines
failed_pipelines = []
(pipelines["pipeline_1"]["status"] == "failed") and failed_pipelines.append("pipeline_1")
(pipelines["pipeline_2"]["status"] == "failed") and failed_pipelines.append("pipeline_2")
(pipelines["pipeline_3"]["status"] == "failed") and failed_pipelines.append("pipeline_3")

# 3. Largest pipeline total
largest_pipeline_total = max(tot1, tot2, tot3)

# 4. Average records per run for each
avg1 = tot1 / len(pipelines["pipeline_1"]["records"])
avg2 = tot2 / len(pipelines["pipeline_2"]["records"])
avg3 = tot3 / len(pipelines["pipeline_3"]["records"])
average_records_per_run = {"pipeline_1": round(avg1, 2), "pipeline_2": round(avg2, 2), "pipeline_3": round(avg3, 2)}

print("Total Processed Records:", total_processed_records)
print("Failed Pipelines:", failed_pipelines)
print("Largest Pipeline Total:", largest_pipeline_total)
print("Average Records Per Run:", average_records_per_run)
```

## Q48. Banking Transactions

```
accounts = {
 "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
 "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
 "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

# 1. Final balances
bal1 = sum(accounts["A101"]["transactions"])
bal2 = sum(accounts["A102"]["transactions"])
bal3 = sum(accounts["A103"]["transactions"])
final_balances = {"A101": bal1, "A102": bal2, "A103": bal3}

# 2. Highest balance
highest_balance = max(bal1, bal2, bal3)

# 3. Total withdrawals
t1, t2, t3 = accounts["A101"]["transactions"], accounts["A102"]["transactions"], accounts["A103"]["transactions"]
w1 = (t1[0] if t1[0] < 0 else 0) + (t1[1] if t1[1] < 0 else 0) + (t1[2] if t1[2] < 0 else 0) + (t1[3] if t1[3] < 0 else 0)
w2 = (t2[0] if t2[0] < 0 else 0) + (t2[1] if t2[1] < 0 else 0) + (t2[2] if t2[2] < 0 else 0)
w3 = (t3[0] if t3[0] < 0 else 0) + (t3[1] if t3[1] < 0 else 0) + (t3[2] if t3[2] < 0 else 0)
total_withdrawals = w1 + w2 + w3

# 4. Largest single deposit
largest_single_deposit = max(max(t1), max(t2), max(t3))

print("Final Balances:", final_balances)
print("Highest Balance:", highest_balance)
print("Total Withdrawals:", total_withdrawals)
print("Largest Single Deposit:", largest_single_deposit)
```

## Q49. Product Recommendations

```
users = {
 "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
 "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
 "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

p_ali, w_ali = users["Ali"]["purchased"], users["Ali"]["wishlist"]
p_sara, w_sara = users["Sara"]["purchased"], users["Sara"]["wishlist"]
p_ahmed, w_ahmed = users["Ahmed"]["purchased"], users["Ahmed"]["wishlist"]

# 1. Products bought by at least two users
bought_by_at_least_two = (p_ali & p_sara) | (p_sara & p_ahmed) | (p_ali & p_ahmed)

# 2. Ali wishlist minus purchases
ali_wishlist_minus_purchases = w_ali - p_ali

# 3. Shared wishlist products
shared_wishlist_products = w_ali & w_sara & w_ahmed

# 4. All products
all_products = p_ali | w_ali | p_sara | w_sara | p_ahmed | w_ahmed

print("Bought by at least two:", bought_by_at_least_two)
print("Ali Wishlist - Purchases:", ali_wishlist_minus_purchases)
print("Shared Wishlist Products:", shared_wishlist_products)
print("All Products:", all_products)
```

## Q50. Final Company Analysis

```
company = {
 "departments": {
 "Engineering": {
 "employees": [
 ("Ali", 120000, {"Python", "AWS", "SQL"}),
 ("Sara", 150000, {"Python", "Databricks", "SQL"})
 ],
 "budget": 500000
 },
 "Analytics": {
 "employees": [
 ("Ahmed", 100000, {"SQL", "Power BI"}),
 ("Zain", 130000, {"Python", "SQL", "Tableau"})
 ],
 "budget": 400000
 }
 }
}

eng_emp = company["departments"]["Engineering"]["employees"]
ana_emp = company["departments"]["Analytics"]["employees"]
eng_budget = company["departments"]["Engineering"]["budget"]
ana_budget = company["departments"]["Analytics"]["budget"]

# 1. Payroll per department
eng_payroll = eng_emp[0][1] + eng_emp[1][1]
ana_payroll = ana_emp[0][1] + ana_emp[1][1]
payroll_per_department = {"Engineering": eng_payroll, "Analytics": ana_payroll}

# 2. Company payroll
company_payroll = eng_payroll + ana_payroll

# 3. Highest-paid employee
highest_paid_employee = max((eng_emp[0][1], eng_emp[0][0]), (eng_emp[1][1], eng_emp[1][0]), (ana_emp[0][1], ana_emp[0][0]), (ana_emp[1][1], ana_emp[1][0]))[1]

# 4. Skills shared by all
skills_shared_by_all = eng_emp[0][2] & eng_emp[1][2] & ana_emp[0][2] & ana_emp[1][2]

# 5. Engineering-only skills
eng_skills = eng_emp[0][2] | eng_emp[1][2]
ana_skills = ana_emp[0][2] | ana_emp[1][2]
engineering_only_skills = eng_skills - ana_skills

# 6. Remaining budget per department
remaining_budget = {"Engineering": eng_budget - eng_payroll, "Analytics": ana_budget - ana_payroll}

# 7. Department with higher average salary
higher_average_dept = "Engineering" if (eng_payroll/len(eng_emp)) > (ana_payroll/len(ana_emp)) else "Analytics"

# 8. Name-to-salary dictionary
name_to_salary = {eng_emp[0][0]: eng_emp[0][1], eng_emp[1][0]: eng_emp[1][1], ana_emp[0][0]: ana_emp[0][1], ana_emp[1][0]: ana_emp[1][1]}

# 9. All unique skills
all_unique_skills = eng_skills | ana_skills

# 10. Is combined budget sufficient?
is_combined_budget_sufficient = (eng_budget + ana_budget) >= company_payroll

print("1. Payroll per Department:", payroll_per_department)
print("2. Company Payroll:", company_payroll)
print("3. Highest Paid Employee:", highest_paid_employee)
print("4. Skills Shared By All:", skills_shared_by_all)
print("5. Engineering-Only Skills:", engineering_only_skills)
print("6. Remaining Budget:", remaining_budget)
print("7. Department with Higher Average:", higher_average_dept)
print("8. Name-to-Salary Dictionary:", name_to_salary)
print("9. All Unique Skills:", all_unique_skills)
print("10. Is Combined Budget Sufficient?:", is_combined_budget_sufficient)
```
