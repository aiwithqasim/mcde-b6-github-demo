Q11. Tuple Indexing

```python
colors = ("red", "green", "blue", "yellow", "purple")

second = colors[1]
last = colors[-1]
middle_three = colors[1:4]

print("Second:", second)
print("Last:", last)
print("Middle Three:", middle_three)
```

Q13. Tuple Concatenation

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

combined = tuple1 + tuple2
reversed_tuple = combined[::-1]

print("Combined:", combined)
print("Reversed:", reversed_tuple)
```

Q15. Extended Tuple Unpacking

```python
data = (100, 200, 300, 400, 500, 600)

first, second, *middle, last = data

middle_list = list(middle)

print("First:", first)
print("Second:", second)
print("Middle:", middle_list)
print("Last:", last)
```

Q16. Modify an Immutable Tuple

```python
numbers = (10, 20, 30, 40, 50)

new_numbers = numbers[:2] + (300,) + numbers[3:]

print("Original:", numbers)
print("New Tuple:", new_numbers)
```

Q17. Nested Tuple Access

```python
students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)

sara_second_score = students[1][1][1]

ali_scores = students[0][1]
ali_average = sum(ali_scores) / len(ali_scores)

ahmed_scores = students[2][1]
ahmed_highest = max(ahmed_scores)

print("Sara Second Score:", sara_second_score)
print("Ali Average:", ali_average)
print("Ahmed Highest:", ahmed_highest)
```

Q18. Tuple Sorting

```python
employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

sorted_employees = sorted(
    employees,
    key=lambda employee: employee[1],
    reverse=True
)

print(sorted_employees)
```

Q19. Tuple Matrix

```python
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])

diagonal_product = diagonal[0] * diagonal[1] * diagonal[2]

row_sums = (
    sum(matrix[0]),
    sum(matrix[1]),
    sum(matrix[2])
)

print("Diagonal:", diagonal)
print("Diagonal Product:", diagonal_product)
print("Row Sums:", row_sums)
```

Q20. Inventory Tuple Challenge

```python
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

laptop_value = records[0][1] * records[0][2]
mouse_value = records[1][1] * records[1][2]
keyboard_value = records[2][1] * records[2][2]

total_inventory_value = laptop_value + mouse_value + keyboard_value

most_expensive = max(records, key=lambda item: item[1])

names = (records[0][0], records[1][0], records[2][0])
reversed_names = names[::-1]

print("Inventory Value:", total_inventory_value)
print("Most Expensive Product:", most_expensive[0])
print("Reversed Names:", reversed_names)
```
