Q41. Student Performance System

```python
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

ali_avg = sum(students[0]["marks"]) / len(students[0]["marks"])
sara_avg = sum(students[1]["marks"]) / len(students[1]["marks"])
ahmed_avg = sum(students[2]["marks"]) / len(students[2]["marks"])

averages = {
    students[0]["name"]: ali_avg,
    students[1]["name"]: sara_avg,
    students[2]["name"]: ahmed_avg
}

top_student = max(averages, key=averages.get)

names = (
    students[0]["name"],
    students[1]["name"],
    students[2]["name"]
)

unique_marks = set(
    students[0]["marks"]
    + students[1]["marks"]
    + students[2]["marks"]
)

print("Averages:", averages)
print("Top Student:", top_student)
print("Names:", names)
print("Unique Marks:", unique_marks)
```

Q45. Course Enrollment

```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

python_students = courses["Python"]
sql_students = courses["SQL"]
aws_students = courses["AWS"]

all_three = python_students & sql_students & aws_students

exactly_two = (
    ((python_students & sql_students) - aws_students)
    | ((python_students & aws_students) - sql_students)
    | ((sql_students & aws_students) - python_students)
)

exactly_one = (
    (python_students - sql_students - aws_students)
    | (sql_students - python_students - aws_students)
    | (aws_students - python_students - sql_students)
)

all_students = python_students | sql_students | aws_students

print("Exactly Two Courses:", exactly_two)
print("All Three Courses:", all_three)
print("Exactly One Course:", exactly_one)
print("Unique Students:", len(all_students))
```

Q46. E-Commerce Transactions

```python
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

transaction1 = (
    transactions[0]["items"][0][1] * transactions[0]["items"][0][2]
    + transactions[0]["items"][1][1] * transactions[0]["items"][1][2]
)

transaction2 = (
    transactions[1]["items"][0][1] * transactions[1]["items"][0][2]
    + transactions[1]["items"][1][1] * transactions[1]["items"][1][2]
)

transaction3 = (
    transactions[2]["items"][0][1] * transactions[2]["items"][0][2]
)

total_revenue = transaction1 + transaction2 + transaction3

ali_spending = transaction1 + transaction3

transaction_totals = {
    transactions[0]["id"]: transaction1,
    transactions[1]["id"]: transaction2,
    transactions[2]["id"]: transaction3
}

largest_transaction = max(transaction_totals, key=transaction_totals.get)

products = {
    transactions[0]["items"][0][0],
    transactions[0]["items"][1][0],
    transactions[1]["items"][0][0],
    transactions[1]["items"][1][0],
    transactions[2]["items"][0][0]
}

print("Total Revenue:", total_revenue)
print("Ali Spending:", ali_spending)
print("Largest Transaction ID:", largest_transaction)
print("Unique Products:", products)
```

Q47. Pipeline Monitoring

```python
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

p1_records = pipelines["pipeline_1"]["records"]
p2_records = pipelines["pipeline_2"]["records"]
p3_records = pipelines["pipeline_3"]["records"]

p1_total = sum(p1_records)
p2_total = sum(p2_records)
p3_total = sum(p3_records)

total_records = p1_total + p2_total + p3_total

failed_pipelines = {
    "pipeline_2"
} if pipelines["pipeline_2"]["status"] == "failed" else set()

pipeline_totals = {
    "pipeline_1": p1_total,
    "pipeline_2": p2_total,
    "pipeline_3": p3_total
}

largest_pipeline = max(pipeline_totals, key=pipeline_totals.get)

print("Total Records:", total_records)
print("Failed Pipelines:", failed_pipelines)
print("Largest Pipeline:", largest_pipeline)
print("Pipeline 1 Average:", p1_total / len(p1_records))
print("Pipeline 2 Average:", p2_total / len(p2_records))
print("Pipeline 3 Average:", p3_total / len(p3_records))
```

Q48. Banking Transactions

```python
accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

ali_balance = sum(accounts["A101"]["transactions"])
sara_balance = sum(accounts["A102"]["transactions"])
ahmed_balance = sum(accounts["A103"]["transactions"])

balances = {
    accounts["A101"]["holder"]: ali_balance,
    accounts["A102"]["holder"]: sara_balance,
    accounts["A103"]["holder"]: ahmed_balance
}

highest_balance = max(balances, key=balances.get)

ali_withdrawals = abs(-2000) + abs(-1000)
sara_withdrawals = abs(-3000) + abs(-2000)
ahmed_withdrawals = abs(-1500)

total_withdrawals = ali_withdrawals + sara_withdrawals + ahmed_withdrawals

largest_deposit = max(5000, 3000, 10000, 7000, 2500)

print("Final Balances:", balances)
print("Highest Balance:", highest_balance)
print("Total Withdrawals:", total_withdrawals)
print("Largest Deposit:", largest_deposit)
```

Q49. Product Recommendations

```python
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

ali_purchases = users["Ali"]["purchased"]
sara_purchases = users["Sara"]["purchased"]
ahmed_purchases = users["Ahmed"]["purchased"]

bought_by_two = (
    (ali_purchases & sara_purchases)
    | (ali_purchases & ahmed_purchases)
    | (sara_purchases & ahmed_purchases)
)

ali_recommendations = users["Ali"]["wishlist"] - ali_purchases

shared_wishlist = (
    (users["Ali"]["wishlist"] & users["Sara"]["wishlist"])
    | (users["Ali"]["wishlist"] & users["Ahmed"]["wishlist"])
    | (users["Sara"]["wishlist"] & users["Ahmed"]["wishlist"])
)

all_products = (
    ali_purchases
    | sara_purchases
    | ahmed_purchases
    | users["Ali"]["wishlist"]
    | users["Sara"]["wishlist"]
    | users["Ahmed"]["wishlist"]
)

print("Bought by at Least Two Users:", bought_by_two)
print("Ali Recommendations:", ali_recommendations)
print("Shared Wishlist:", shared_wishlist)
print("All Products:", all_products)
```

Q50. Final Company Analysis

```python
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

engineering = company["departments"]["Engineering"]
analytics = company["departments"]["Analytics"]

eng_employees = engineering["employees"]
ana_employees = analytics["employees"]

# 1. Payroll per department
eng_payroll = eng_employees[0][1] + eng_employees[1][1]
ana_payroll = ana_employees[0][1] + ana_employees[1][1]

# 2. Company payroll
company_payroll = eng_payroll + ana_payroll

# 3. Highest-paid employee
all_salaries = {
    eng_employees[0][0]: eng_employees[0][1],
    eng_employees[1][0]: eng_employees[1][1],
    ana_employees[0][0]: ana_employees[0][1],
    ana_employees[1][0]: ana_employees[1][1]
}

highest_paid = max(all_salaries, key=all_salaries.get)

# 4. Skills shared by all employees
ali_skills = eng_employees[0][2]
sara_skills = eng_employees[1][2]
ahmed_skills = ana_employees[0][2]
zain_skills = ana_employees[1][2]

shared_skills = ali_skills & sara_skills & ahmed_skills & zain_skills

# 5. Engineering-only skills
eng_skills = ali_skills | sara_skills
ana_skills = ahmed_skills | zain_skills

engineering_only = eng_skills - ana_skills

# 6. Remaining budget
eng_remaining = engineering["budget"] - eng_payroll
ana_remaining = analytics["budget"] - ana_payroll

# 7. Department with higher average salary
eng_average = eng_payroll / len(eng_employees)
ana_average = ana_payroll / len(ana_employees)

higher_average = (
    "Engineering" if eng_average > ana_average else "Analytics"
)

# 8. Name-to-salary dictionary
name_salary = all_salaries

# 9. All unique skills
all_skills = eng_skills | ana_skills

# 10. Is combined budget sufficient?
total_budget = engineering["budget"] + analytics["budget"]
budget_sufficient = total_budget >= company_payroll

print("Engineering Payroll:", eng_payroll)
print("Analytics Payroll:", ana_payroll)
print("Company Payroll:", company_payroll)
print("Highest-Paid Employee:", highest_paid)
print("Shared Skills:", shared_skills)
print("Engineering-Only Skills:", engineering_only)
print("Engineering Remaining Budget:", eng_remaining)
print("Analytics Remaining Budget:", ana_remaining)
print("Higher Average Salary:", higher_average)
print("Name to Salary:", name_salary)
print("All Unique Skills:", all_skills)
print("Combined Budget Sufficient:", budget_sufficient)
```
