Q17. Nested Tuple Access
students = (
    ("Ali", (85, 90, 88)),

    ("Ahmed", (78, 84, 80))
)
Extract Sara second score, Ali average, and Ahmed highest score

```python
students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)
sara_s =  students[1][1][1]
ali_a = sum(students[0][1])/len(students[0][1])
ahmed_h = max(students[2][1])
print("Sara's second score:", sara_s)      
print("Ali's average:", ali_a)            
print("Ahmed's highest score:", ahmed_h)
```

Q18. Tuple Sorting
employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)
Sort by salary descending without loops, comprehensions, or lambda

```python
employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

e = list(employees)

pairs = [
    (e[0][1], e[0][0]),
    (e[1][1], e[1][0]),
    (e[2][1], e[2][0])
]

pairs.sort(reverse=True)   

result = (
    (pairs[0][1], pairs[0][0]),
    (pairs[1][1], pairs[1][0]),
    (pairs[2][1], pairs[2][0])
)

print(result)
```

Q19. Tuple Matrix
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))
Extract diagonal, calculate its product, and build a tuple of row sums

```python
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])
print("Diagonal:", diagonal)            

product = diagonal[0] * diagonal[1] * diagonal[2]
print("Product:", product)              

row_sums = (sum(matrix[0]), sum(matrix[1]), sum(matrix[2]))
print("Row sums:", row_sums)            
```

Q20. Inventory Tuple Challenge
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)
Calculate inventory value, find most expensive product, and build reversed tuple of names

```python
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

v1 = records[0][1] * records[0][2]      
v2 = records[1][1] * records[1][2]      
v3 = records[2][1] * records[2][2]      
total_value = v1 + v2 + v3
print("Inventory value:", total_value)  

prices = (records[0][1], records[1][1], records[2][1])
highest = max(prices)
expensive = records[prices.index(highest)][0]
print("Most expensive:", expensive)     

names = (records[0][0], records[1][0], records[2][0])
reversed_names = names[::-1]
print("Reversed names:", reversed_names)  
```
