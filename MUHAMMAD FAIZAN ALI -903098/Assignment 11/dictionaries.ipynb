Q33. Dictionary Methods

product = {"id": 101, "name": "Laptop", "price": 120000, "stock": 5}
Extract keys, values, items; remove stock and retain its value.

```python
product = {"id": 101, "name": "Laptop", "price": 120000, "stock": 5}

print("Keys:", list(product.keys()))
print("Values:", list(product.values()))
print("Items:", list(product.items()))

removed_stock = product.pop("stock")     
print("Removed stock:", removed_stock)
print("Product now:", product)
```

Q35. Student Marks

marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}
Find highest, lowest, average, and top-scoring student without loops or lambda.

```python
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}

highest = max(marks.values())
lowest = min(marks.values())
average = sum(marks.values()) / len(marks)
top_student = max(marks, key=marks.get)    

print("Highest:", highest)
print("Lowest:", lowest)
print("Average:", average)
print("Top student:", top_student)
```

Q36. Nested Employee Records

employees = {"E101": {...Ali, 80000}, "E102": {...Sara, 120000}, "E103": {...Ahmed, 95000}}
Extract Sara salary, increase Ali salary 15%, and calculate total payroll.

```python
employees = {
    "E101": {"name": "Ali", "salary": 80000},
    "E102": {"name": "Sara", "salary": 120000},
    "E103": {"name": "Ahmed", "salary": 95000}
}

print("Sara's salary:", employees["E102"]["salary"])

employees["E101"]["salary"] = employees["E101"]["salary"] * 1.15
print("Ali's new salary:", employees["E101"]["salary"])

e1, e2, e3 = employees.values()
total_payroll = e1["salary"] + e2["salary"] + e3["salary"]
print("Total payroll:", total_payroll)
```

Q37. Dictionary Inversion

data = {"Python": "Ali", "SQL": "Sara", "AWS": "Ahmed"}
Create inverse dictionary without loops or comprehensions.

```python
data = {"Python": "Ali", "SQL": "Sara", "AWS": "Ahmed"}

inverse = dict(zip(data.values(), data.keys()))
print(inverse)
```

Q38. Order Calculation

order = {"laptop": {...}, "mouse": {...}, "keyboard": {...}}
Calculate subtotals, grand total, and amount after 10% discount.

```python
order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}

laptop, mouse, keyboard = order.values()

subtotals = dict(zip(order.keys(), (
    laptop["price"] * laptop["quantity"],
    mouse["price"] * mouse["quantity"],
    keyboard["price"] * keyboard["quantity"]
)))

grand_total = sum(subtotals.values())
after_discount = grand_total * 0.90

print("Subtotals:", subtotals)
print("Grand total:", grand_total)
print("After 10% discount:", after_discount)
```

Q39. Employee Performance

employees = {"Ali": {python, sql, aws}, "Sara": {...}, "Ahmed": {...}}
Calculate each average, top performer, and highest SQL score.

```python
employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

ali, sara, ahmed = employees.values()

averages = dict(zip(employees.keys(), (
    sum(ali.values()) / len(ali),
    sum(sara.values()) / len(sara),
    sum(ahmed.values()) / len(ahmed)
)))
top_performer = max(averages, key=averages.get)

sql_scores = dict(zip(employees.keys(), (ali["sql"], sara["sql"], ahmed["sql"])))
highest_sql = max(sql_scores.values())
highest_sql_name = max(sql_scores, key=sql_scores.get)

print("Averages:", averages)
print("Top performer:", top_performer)
print("Highest SQL score:", highest_sql, "by", highest_sql_name)
```

Q40. Inventory Management

inventory = {"P101": {Laptop}, "P102": {Mouse}, "P103": {Keyboard}}
Calculate total value, most valuable item by price x stock, and update stock after selling two laptops and five mice.

```python
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

p1, p2, p3 = inventory.values()

item_values = dict(zip(inventory.keys(), (
    p1["price"] * p1["stock"],
    p2["price"] * p2["stock"],
    p3["price"] * p3["stock"]
)))

total_value = sum(item_values.values())
most_valuable_id = max(item_values, key=item_values.get)

print("Item values:", item_values)
print("Total inventory value:", total_value)
print("Most valuable item:", inventory[most_valuable_id]["name"])

inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5
print("Updated stock:")
print(inventory)
```
