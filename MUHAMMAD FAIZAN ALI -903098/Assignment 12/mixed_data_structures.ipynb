Q41. Student Performance System

students = [{id, name, marks (tuple)}, ...]
Calculate each average, top student, tuple of names, and set of unique marks.

```python
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

s1, s2, s3 = students
names = (s1["name"], s2["name"], s3["name"])

averages = dict(zip(names, (
    sum(s1["marks"]) / len(s1["marks"]),
    sum(s2["marks"]) / len(s2["marks"]),
    sum(s3["marks"]) / len(s3["marks"])
)))

top_student = max(averages, key=averages.get)
unique_marks = set(s1["marks"]) | set(s2["marks"]) | set(s3["marks"])

print("Averages:", averages)
print("Top student:", top_student)
print("Names:", names)
print("Unique marks:", unique_marks)
```

Q42. Employee Skills

employees = {"Ali": {...}, "Sara": {...}, "Ahmed": {...}}
Find common skills, Ali-only skills, all team skills, and employees meeting Python + SQL requirements.

```python
employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

ali, sara, ahmed = employees.values()

common_skills = ali & sara & ahmed
ali_only = ali - (sara | ahmed)
all_skills = ali | sara | ahmed

required = {"Python", "SQL"}
meets = dict(zip(employees.keys(), (
    required.issubset(ali),
    required.issubset(sara),
    required.issubset(ahmed)
)))

qualified = sorted(employees, key=meets.get, reverse=True)[:sum(meets.values())]

print("Common skills:", common_skills)
print("Ali-only skills:", ali_only)
print("All team skills:", all_skills)
print("Meets Python + SQL:", meets)
print("Qualified employees:", qualified)
```

Q43. Sales Data

sales = [("Laptop", 120000, 2), ("Mouse", 1500, 5), ("Keyboard", 3500, 3)]
Calculate revenue, top-revenue product, name-to-revenue dictionary, and set of prices.

```python
sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

names, prices, quantities = zip(*sales)

revenues = (
    prices[0] * quantities[0],
    prices[1] * quantities[1],
    prices[2] * quantities[2]
)

total_revenue = sum(revenues)
name_to_revenue = dict(zip(names, revenues))
top_product = max(name_to_revenue, key=name_to_revenue.get)
unique_prices = set(prices)

print("Revenues:", revenues)
print("Total revenue:", total_revenue)
print("Top-revenue product:", top_product)
print("Name to revenue:", name_to_revenue)
print("Set of prices:", unique_prices)
```

Q45. Course Enrollment

courses = {"Python": {...}, "SQL": {...}, "AWS": {...}}
Find students in exactly two, all three, and exactly one course; count unique students.

```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

py, sql, aws = courses.values()

all_three = py & sql & aws
at_least_two = (py & sql) | (py & aws) | (sql & aws)

exactly_two = at_least_two - all_three
exactly_one = (py ^ sql ^ aws) - all_three
unique_count = len(py | sql | aws)

print("Exactly two courses:", exactly_two)
print("All three courses:", all_three)
print("Exactly one course:", exactly_one)
print("Unique students:", unique_count)
```

Q46. E-Commerce Transactions

transactions = [{id, customer, items: [(name, price, qty), ...]}, ...]
Calculate total revenue, Ali spending, largest transaction, and unique product names.

```python
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

t1, t2, t3 = transactions
i1, i2, i3 = t1["items"], t2["items"], t3["items"]

# total of each transaction (price * qty for every item)
total1 = i1[0][1] * i1[0][2] + i1[1][1] * i1[1][2]
total2 = i2[0][1] * i2[0][2] + i2[1][1] * i2[1][2]
total3 = i3[0][1] * i3[0][2]

total_revenue = total1 + total2 + total3

ali_spending = (
    total1 * (t1["customer"] == "Ali")
    + total2 * (t2["customer"] == "Ali")
    + total3 * (t3["customer"] == "Ali")
)

by_id = dict(zip((t1["id"], t2["id"], t3["id"]), (total1, total2, total3)))
largest_id = max(by_id, key=by_id.get)

unique_products = {i1[0][0], i1[1][0], i2[0][0], i2[1][0], i3[0][0]}

print("Transaction totals:", by_id)
print("Total revenue:", total_revenue)
print("Ali spending:", ali_spending)
print("Largest transaction:", largest_id, "with", by_id[largest_id])
print("Unique products:", unique_products)
```

Q47. Pipeline Monitoring

pipelines = {"pipeline_1": {status, records (tuple)}, ...}
Find total processed records, failed pipelines, largest pipeline total, and average records per run for each.

```python
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

names = tuple(pipelines)
p1, p2, p3 = pipelines.values()

totals = dict(zip(names, (sum(p1["records"]), sum(p2["records"]), sum(p3["records"]))))
total_processed = sum(totals.values())

is_failed = dict(zip(names, (
    p1["status"] == "failed",
    p2["status"] == "failed",
    p3["status"] == "failed"
)))

failed = sorted(names, key=is_failed.get, reverse=True)[:sum(is_failed.values())]

largest = max(totals, key=totals.get)

averages = dict(zip(names, (
    sum(p1["records"]) / len(p1["records"]),
    sum(p2["records"]) / len(p2["records"]),
    sum(p3["records"]) / len(p3["records"])
)))

print("Totals:", totals)
print("Total processed records:", total_processed)
print("Failed pipelines:", failed)
print("Largest pipeline:", largest, "with", totals[largest])
print("Average records per run:", averages)
```

Q48. Banking Transactions

accounts = {"A101": {holder, transactions}, "A102": {...}, "A103": {...}}
Assuming zero starting balances, calculate final balances, highest balance, total withdrawals, and largest single deposit.

```python
accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

ids = tuple(accounts)
a1, a2, a3 = accounts.values()
t1, t2, t3 = a1["transactions"], a2["transactions"], a3["transactions"]

balances = dict(zip(ids, (sum(t1), sum(t2), sum(t3))))
highest_id = max(balances, key=balances.get)

sorted_tx = sorted(t1 + t2 + t3 + [0])
first_zero = sorted_tx.index(0)

total_withdrawals = -sum(sorted_tx[:first_zero])
largest_deposit = max(t1 + t2 + t3)

print("Final balances:", balances)
print("Highest balance:", highest_id, "(" + accounts[highest_id]["holder"] + ")", balances[highest_id])
print("Total withdrawals:", total_withdrawals)
print("Largest single deposit:", largest_deposit)
```

Q49. Product Recommendations

users = {"Ali": {purchased, wishlist}, "Sara": {...}, "Ahmed": {...}}
Find products bought by at least two users, Ali wishlist minus purchases, shared wishlist products, and all products.

```python
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

ali, sara, ahmed = users.values()

bought_by_two = (
    (ali["purchased"] & sara["purchased"])
    | (ali["purchased"] & ahmed["purchased"])
    | (sara["purchased"] & ahmed["purchased"])
)

ali_wishlist_left = ali["wishlist"] - ali["purchased"]

shared_wishlist = (
    (ali["wishlist"] & sara["wishlist"])
    | (ali["wishlist"] & ahmed["wishlist"])
    | (sara["wishlist"] & ahmed["wishlist"])
)

all_products = set.union(
    ali["purchased"], ali["wishlist"],
    sara["purchased"], sara["wishlist"],
    ahmed["purchased"], ahmed["wishlist"]
)

print("Bought by at least two users:", bought_by_two)
print("Ali wishlist minus purchases:", ali_wishlist_left)
print("Shared wishlist products:", shared_wishlist)
print("All products:", all_products)
```

Q50. Final Company Analysis

company = {"departments": {"Engineering": {employees, budget}, "Analytics": {...}}}
10 tasks: payroll per department, company payroll, highest-paid employee, skills shared by all, Engineering-only skills, remaining budget, department with higher average salary, name-to-salary dictionary, all unique skills, is combined budget sufficient?

```python
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

eng = company["departments"]["Engineering"]
ana = company["departments"]["Analytics"]

eng_names, eng_salaries, eng_skills = zip(*eng["employees"])
ana_names, ana_salaries, ana_skills = zip(*ana["employees"])

eng_payroll = sum(eng_salaries)
ana_payroll = sum(ana_salaries)
print("1. Payroll per department:", {"Engineering": eng_payroll, "Analytics": ana_payroll})

company_payroll = eng_payroll + ana_payroll
print("2. Company payroll:", company_payroll)

all_employees = eng["employees"] + ana["employees"]
all_names, all_salaries, all_skills = zip(*all_employees)

# highest salary, then use its position to get the matching name
top_salary = max(all_salaries)
top_name = all_names[all_salaries.index(top_salary)]
print("3. Highest-paid employee:", top_name, top_salary)

shared_skills = set.intersection(*all_skills)
print("4. Skills shared by all:", shared_skills)

eng_only = set.union(*eng_skills) - set.union(*ana_skills)
print("5. Engineering-only skills:", eng_only)

remaining = {
    "Engineering": eng["budget"] - eng_payroll,
    "Analytics": ana["budget"] - ana_payroll
}
print("6. Remaining budget:", remaining)

averages = {
    "Engineering": eng_payroll / len(eng_salaries),
    "Analytics": ana_payroll / len(ana_salaries)
}
higher_avg = max(averages, key=averages.get)
print("7. Higher average salary:", higher_avg, averages)

name_to_salary = dict(zip(all_names, all_salaries))
print("8. Name to salary:", name_to_salary)

unique_skills = set.union(*all_skills)
print("9. All unique skills:", unique_skills)

combined_budget = eng["budget"] + ana["budget"]
print("10. Combined budget sufficient:", combined_budget >= company_payroll,
      "(budget", combined_budget, "vs payroll", company_payroll, ")")
```
