## Q1. Accessing List Elements


```python
numbers = [10, 20, 30, 40, 50, 60, 70]
first = numbers[0]
middle = numbers[len(numbers) // 2]
last = numbers[-1]
print('First:', first)
print('Middle:', middle)
print('Last:', last)
```

## Q2. List Slicing


```python
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
first_five = numbers[:5]
last_four = numbers[-4:]
second_reversed = numbers[::-2]
print('First five:', first_five)
print('Last four:', last_four)
print('Every second in reverse:', second_reversed)
```

## Q3. List Modification


```python
fruits = ['apple', 'banana', 'cherry', 'mango']
fruits[fruits.index('banana')] = 'orange'
fruits.insert(2, 'grapes')
fruits.remove('cherry')
print('Updated fruits:', fruits)
```

## Q4. Reverse Without reverse()

Reverse using slicing while preserving the original.

```python
numbers = [12, 45, 23, 67, 89, 34]
reversed_numbers = numbers[::-1]
print('Reversed:', reversed_numbers)
print('Original:', numbers)
```

## Q5. Second Largest Unique Number

Find the second-largest distinct value.

```python
numbers = [45, 12, 89, 34, 89, 67, 45, 23]
second_largest = sorted(set(numbers))[-2]
print('Second largest unique number:', second_largest)
```

## Q6. Rotate a List



```python
numbers = [10, 20, 30, 40, 50, 60]
k = 2
rotated = numbers[-k:] + numbers[:-k]
print('Rotated:', rotated)
```

## Q7. Remove Duplicates, Preserve Order

Keep the first occurrence of each number.

```python
numbers = [10, 20, 10, 30, 20, 40, 30, 50]
unique_in_order = list(dict.fromkeys(numbers))
print('Without duplicates:', unique_in_order)
```

## Q8. Flatten a Nested List

Flatten the given nested list using list operations.

```python
data = [[10, 20], [30, 40], [50, 60]]
flattened = sum(data, [])
print('Flattened:', flattened)
```

## Q9. Extract Data from a Matrix

Extract both diagonals and middle column; calculate their sums.

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

main_diagonal = [matrix[0][0], matrix[1][1], matrix[2][2]]
secondary_diagonal = [matrix[0][-1], matrix[1][-2], matrix[2][-3]]
mid = len(matrix[0]) // 2
middle_column = [matrix[0][mid], matrix[1][mid], matrix[2][mid]]
print('Main diagonal:', main_diagonal, '| Sum:', sum(main_diagonal))
print('Secondary diagonal:', secondary_diagonal, '| Sum:', sum(secondary_diagonal))
print('Middle column:', middle_column, '| Sum:', sum(middle_column))
```

## Q10. List Transformation Challenge

Three largest unique values ascending, followed by two smallest unique values descending.

```python
data = [15, 5, 25, 10, 35, 20, 30]
unique_sorted = sorted(set(data))
result = unique_sorted[-3:] + unique_sorted[:2][::-1]
print('Transformed:', result)
```
