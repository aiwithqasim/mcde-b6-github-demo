### Q22. Set Operations

### Q25. Subset and Superset

```python
required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}
print(required_skills <= candidate_skills)
print(candidate_skills >= required_skills)
```

### Q27. Three-Set Analysis

```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}
print(python & sql & aws)
print((python & sql) - aws)
print((python & sql) | (python & aws) | (sql & aws))
```

### Q28. Transaction Analysis

```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}
all_days = day1 & day2 & day3
one_day = (day1 - day2 - day3) | (day2 - day1 - day3) | (day3 - day1 - day2)
two_or_more = (day1 & day2) | (day1 & day3) | (day2 & day3)
print(all_days)
print(one_day)
print(two_or_more)
```

### Q29. Access Control

```python
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}
print(required_permissions - user_permissions)
print(user_permissions - required_permissions)
print(required_permissions <= user_permissions)
```

### Q30. Customer Segmentation

```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}
one = (website - mobile - store) | (mobile - website - store) | (store - website - mobile)
all_three = website & mobile & store
two = ((website & mobile) | (website & store) | (mobile & store)) - all_three
print(one)
print(two)
print(all_three)
```
