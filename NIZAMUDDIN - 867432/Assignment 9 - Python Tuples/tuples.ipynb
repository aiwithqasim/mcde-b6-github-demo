```python
from operator import itemgetter

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)
print(tuple(sorted(employees, key=itemgetter(1), reverse=True)))
```

```python
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))
diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])
product = diagonal[0] * diagonal[1] * diagonal[2]
row_sums = (sum(matrix[0]), sum(matrix[1]), sum(matrix[2]))
print(diagonal)
print(product)
print(row_sums)
```

```python
from operator import itemgetter

records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)
value = records[0][1] * records[0][2] + records[1][1] * records[1][2] + records[2][1] * records[2][2]
expensive = max(records, key=itemgetter(1))
names = (records[0][0], records[1][0], records[2][0])[::-1]
print(value)
print(expensive[0])
print(names)
```
