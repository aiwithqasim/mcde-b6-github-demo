```python
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]
a = sum(students[0]["marks"]) / len(students[0]["marks"])
b = sum(students[1]["marks"]) / len(students[1]["marks"])
c = sum(students[2]["marks"]) / len(students[2]["marks"])
averages = dict(zip((students[0]["name"], students[1]["name"], students[2]["name"]), (a, b, c)))
top = max(averages, key=averages.get)
names = tuple(averages)
unique_marks = set(students[0]["marks"] + students[1]["marks"] + students[2]["marks"])
print(averages, top, names, unique_marks)
```

```python
employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}
common = employees["Ali"] & employees["Sara"] & employees["Ahmed"]
ali_only = employees["Ali"] - employees["Sara"] - employees["Ahmed"]
all_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]
required = {"Python", "SQL"}
qualified = ({"Ali"} if required <= employees["Ali"] else set()) | ({"Sara"} if required <= employees["Sara"] else set()) | ({"Ahmed"} if required <= employees["Ahmed"] else set())
print(common, ali_only, all_skills, qualified)
```

```python
sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]
revenues = {sales[0][0]: sales[0][1] * sales[0][2], sales[1][0]: sales[1][1] * sales[1][2], sales[2][0]: sales[2][1] * sales[2][2]}
total = sum(revenues.values())
top = max(revenues, key=revenues.get)
prices = {sales[0][1], sales[1][1], sales[2][1]}
print(total, top, revenues, prices)
```

```python
customers = {
    "C101": {"name": "Ali", "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara", "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}
spending = {"C101": sum(customers["C101"]["orders"]), "C102": sum(customers["C102"]["orders"]), "C103": sum(customers["C103"]["orders"])}
top = max(spending, key=spending.get)
totals = tuple(spending.values())
print(spending, customers[top]["name"], totals)
```

```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}
a, b, c = courses.values()
all_three = a & b & c
exactly_two = ((a & b) | (a & c) | (b & c)) - all_three
exactly_one = (a | b | c) - ((a & b) | (a & c) | (b & c))
unique_count = len(a | b | c)
print(exactly_two, all_three, exactly_one, unique_count)
```

```python
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]
a = transactions[0]["items"][0][1] * transactions[0]["items"][0][2] + transactions[0]["items"][1][1] * transactions[0]["items"][1][2]
b = transactions[1]["items"][0][1] * transactions[1]["items"][0][2] + transactions[1]["items"][1][1] * transactions[1]["items"][1][2]
c = transactions[2]["items"][0][1] * transactions[2]["items"][0][2]
values = {transactions[0]["id"]: a, transactions[1]["id"]: b, transactions[2]["id"]: c}
total = sum(values.values())
ali_spending = (a if transactions[0]["customer"] == "Ali" else 0) + (b if transactions[1]["customer"] == "Ali" else 0) + (c if transactions[2]["customer"] == "Ali" else 0)
largest = max(values, key=values.get)
products = {transactions[0]["items"][0][0], transactions[0]["items"][1][0], transactions[1]["items"][0][0], transactions[1]["items"][1][0], transactions[2]["items"][0][0]}
print(total, ali_spending, largest, products)
```

```python
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}
totals = {"pipeline_1": sum(pipelines["pipeline_1"]["records"]), "pipeline_2": sum(pipelines["pipeline_2"]["records"]), "pipeline_3": sum(pipelines["pipeline_3"]["records"])}
failed = (set(["pipeline_1"]) if pipelines["pipeline_1"]["status"] == "failed" else set()) | (set(["pipeline_2"]) if pipelines["pipeline_2"]["status"] == "failed" else set()) | (set(["pipeline_3"]) if pipelines["pipeline_3"]["status"] == "failed" else set())
averages = {"pipeline_1": totals["pipeline_1"] / len(pipelines["pipeline_1"]["records"]), "pipeline_2": totals["pipeline_2"] / len(pipelines["pipeline_2"]["records"]), "pipeline_3": totals["pipeline_3"] / len(pipelines["pipeline_3"]["records"])}
print(sum(totals.values()), failed, max(totals, key=totals.get), averages)
```

```python
accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}
a = accounts["A101"]["transactions"]
b = accounts["A102"]["transactions"]
c = accounts["A103"]["transactions"]
balances = {"A101": sum(a), "A102": sum(b), "A103": sum(c)}
all_transactions = a + b + c
from bisect import bisect_left
transactions_sorted = sorted(all_transactions)
withdrawals = -sum(transactions_sorted[:bisect_left(transactions_sorted, 0)])
print(balances, max(balances, key=balances.get), withdrawals, max(all_transactions))
```

```python
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}
a, b, c = users["Ali"], users["Sara"], users["Ahmed"]
bought_twice = (a["purchased"] & b["purchased"]) | (a["purchased"] & c["purchased"]) | (b["purchased"] & c["purchased"])
ali_wishlist = a["wishlist"] - a["purchased"]
shared_wishlist = (a["wishlist"] & b["wishlist"]) | (a["wishlist"] & c["wishlist"]) | (b["wishlist"] & c["wishlist"])
all_products = a["purchased"] | a["wishlist"] | b["purchased"] | b["wishlist"] | c["purchased"] | c["wishlist"]
print(bought_twice, ali_wishlist, shared_wishlist, all_products)
```

```python
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}
e = company["departments"]["Engineering"]
a = company["departments"]["Analytics"]
e1, e2 = e["employees"]
a1, a2 = a["employees"]
payroll = {"Engineering": e1[1] + e2[1], "Analytics": a1[1] + a2[1]}
company_payroll = sum(payroll.values())
salaries = {e1[0]: e1[1], e2[0]: e2[1], a1[0]: a1[1], a2[0]: a2[1]}
highest_paid = max(salaries, key=salaries.get)
shared_skills = e1[2] & e2[2] & a1[2] & a2[2]
engineering_only = (e1[2] | e2[2]) - (a1[2] | a2[2])
remaining = {"Engineering": e["budget"] - payroll["Engineering"], "Analytics": a["budget"] - payroll["Analytics"]}
averages = {"Engineering": payroll["Engineering"] / len(e["employees"]), "Analytics": payroll["Analytics"] / len(a["employees"])}
higher_average = max(averages, key=averages.get)
all_skills = e1[2] | e2[2] | a1[2] | a2[2]
budget_sufficient = e["budget"] + a["budget"] >= company_payroll
print("Payroll:", payroll)
print("Company payroll:", company_payroll)
print("Highest paid:", highest_paid)
print("Shared skills:", shared_skills)
print("Engineering only:", engineering_only)
print("Remaining budget:", remaining)
print("Higher average:", higher_average)
print("Salaries:", salaries)
print("All skills:", all_skills)
print("Budget sufficient:", budget_sufficient)
```
