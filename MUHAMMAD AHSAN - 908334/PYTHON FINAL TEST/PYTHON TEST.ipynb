```python

transactions = [
    {"id": 1, "customer": "Ali", "category": "Food", "amount": 500, "status": "success"},
    {"id": 2, "customer": "Ali", "category": "Travel", "amount": 200, "status": "failed"},
    {"id": 3, "customer": "Sara", "category": "Food", "amount": 700, "status": "success"},
    {"id": 4, "customer": "Ali", "category": "Food", "amount": 300, "status": "success"},
    {"id": 5, "customer": "Ahmed", "category": "Shopping", "amount": 400, "status": "failed"},
    {"id": 3, "customer": "Sara", "category": "Food", "amount": 700, "status": "success"}
]

seen_ids = set()
customer_spending = {}
category_spending = {}

for transaction in transactions:

    if transaction["id"] in seen_ids:
        continue

    seen_ids.add(transaction["id"])

    customer = transaction["customer"]

    if customer not in customer_spending:
        customer_spending[customer] = 0
        category_spending[customer] = {}

    if transaction["status"] == "success":

        amount = transaction["amount"]
        category = transaction["category"]

        customer_spending[customer] += amount

        if category not in category_spending[customer]:
            category_spending[customer][category] = 0

        category_spending[customer][category] += amount


report = []

for customer in customer_spending:

    top_category = None

    for category in category_spending[customer]:

        if top_category == None:
            top_category = category

        elif category_spending[customer][category] > category_spending[customer][top_category]:
            top_category = category

    report.append({
        "customer": customer,
        "total_spending": customer_spending[customer],
        "top_category": top_category
    })


report.sort(key=lambda x: (-x["total_spending"], x["customer"]))

rank = 1

for customer in report:
    customer["rank"] = rank
    rank += 1

for customer in report:
    print(customer)
```

```python

def longest_unique_window(text):

    char_index = {}
    start = 0
    longest = ""
    longest_start = 0
    longest_length = 0

    for i in range(len(text)):

        char = text[i]

        if char in char_index and char_index[char] >= start:
            start = char_index[char] + 1

        char_index[char] = i

        current_length = i - start + 1

        if current_length > longest_length:
            longest_length = current_length
            longest_start = start
            longest = text[start:i + 1]

    return longest, longest_start, longest_length


text = input("Enter text: ")

result = longest_unique_window(text)

print(result)
```

```python

def find_paths(data, target):

    paths = []

    if isinstance(data, dict):

        for key, value in data.items():

            if value == target and type(value) == type(target):
                paths.append(str(key))

            else:
                result = find_paths(value, target)

                for path in result:
                    paths.append(str(key) + "." + path)

    elif isinstance(data, list):

        for index in range(len(data)):

            value = data[index]

            if value == target and type(value) == type(target):
                paths.append(f"[{index}]")

            else:
                result = find_paths(value, target)

                for path in result:
                    paths.append(f"[{index}]" + path)

    elif data == target and type(data) == type(target):
        paths.append("")

    return paths


data = {
    "a": [{"x": 5}, 5],
    "b": {"y": 5}
}

target = 5

result = find_paths(data, target)

print(result)
```

```python

def execution_planner(tasks, dependencies):

    graph = {}
    in_degree = {}

    for task in tasks:
        graph[task] = []
        in_degree[task] = 0

    for A, B in dependencies:

        if A not in graph or B not in graph:
            raise ValueError("Unknown task name found")

        graph[A].append(B)
        in_degree[B] += 1

    available = []

    for task in tasks:
        if in_degree[task] == 0:
            available.append(task)

    available.sort()

    order = []

    while available:

        current = available.pop(0)
        order.append(current)

        for next_task in graph[current]:

            in_degree[next_task] -= 1

            if in_degree[next_task] == 0:
                available.append(next_task)

        available.sort()

    if len(order) != len(tasks):
        raise ValueError("Cycle detected in dependencies")

    return order


tasks = ["build", "test", "deploy"]

dependencies = [
    ("build", "test"),
    ("test", "deploy")
]

result = execution_planner(tasks, dependencies)

print(result)
```

```python

def calculate(expression):

    def parse_expression():
        nonlocal position

        result = parse_term()

        while position < len(tokens):

            operator = tokens[position]

            if operator == "+":
                position += 1
                result += parse_term()

            elif operator == "-":
                position += 1
                result -= parse_term()

            else:
                break

        return result

    def parse_term():
        nonlocal position

        result = parse_factor()

        while position < len(tokens):

            operator = tokens[position]

            if operator == "*":
                position += 1
                result *= parse_factor()

            elif operator == "/":
                position += 1
                divisor = parse_factor()

                if divisor == 0:
                    raise ZeroDivisionError("Division by zero")

                result /= divisor

            else:
                break

        return result

    def parse_factor():
        nonlocal position

        if position >= len(tokens):
            raise ValueError("Invalid expression")

        token = tokens[position]

        if token == "-":
            position += 1
            return -parse_factor()

        if token == "+":
            position += 1
            return parse_factor()

        if token == "(":
            position += 1
            result = parse_expression()

            if position >= len(tokens) or tokens[position] != ")":
                raise ValueError("Missing closing parenthesis")

            position += 1
            return result

        if token == ")":
            raise ValueError("Unexpected closing parenthesis")

        position += 1
        return int(token)

    tokens = []
    number = ""

    for char in expression:

        if char.isdigit():
            number += char

        else:
            if number:
                tokens.append(number)
                number = ""

            if char in "+-*/()":
                tokens.append(char)

            elif not char.isspace():
                raise ValueError("Invalid character")

    if number:
        tokens.append(number)

    if not tokens:
        raise ValueError("Empty expression")

    position = 0

    result = parse_expression()

    if position != len(tokens):
        raise ValueError("Invalid expression")

    return result


expression = input("Enter expression: ")

try:
    result = calculate(expression)
    print(result)

except (ValueError, ZeroDivisionError) as error:
    print("Error:", error)
```

```python

from dataclasses import dataclass
from itertools import count


@dataclass(frozen=True)
class Transaction:
    transaction_id: int
    action: str
    amount: float
    resulting_balance: float


class Account:

    transaction_counter = count(1)

    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.__balance = balance
        self.__transactions = []

    @property
    def balance(self):
        return self.__balance

    @property
    def statement(self):
        return tuple(self.__transactions)

    def deposit(self, amount):

        if amount <= 0:
            raise ValueError("Amount must be positive")

        self.__balance += amount

        transaction = Transaction(
            next(self.transaction_counter),
            "Deposit",
            amount,
            self.__balance
        )

        self.__transactions.append(transaction)

    def withdraw(self, amount):

        if amount <= 0:
            raise ValueError("Amount must be positive")

        if amount > self.__balance:
            raise ValueError("Insufficient funds")

        self.__balance -= amount

        transaction = Transaction(
            next(self.transaction_counter),
            "Withdrawal",
            amount,
            self.__balance
        )

        self.__transactions.append(transaction)


class Bank:

    def __init__(self):
        self.accounts = {}

    def create_account(self, account_number, balance=0):

        if account_number in self.accounts:
            raise ValueError("Account already exists")

        if balance < 0:
            raise ValueError("Initial balance cannot be negative")

        self.accounts[account_number] = Account(account_number, balance)

    def get_account(self, account_number):

        if account_number not in self.accounts:
            raise ValueError("Account does not exist")

        return self.accounts[account_number]

    def deposit(self, account_number, amount):

        account = self.get_account(account_number)
        account.deposit(amount)

    def withdraw(self, account_number, amount):

        account = self.get_account(account_number)
        account.withdraw(amount)

    def transfer(self, from_account, to_account, amount):

        if amount <= 0:
            raise ValueError("Amount must be positive")

        sender = self.get_account(from_account)
        receiver = self.get_account(to_account)

        if sender.balance < amount:
            raise ValueError("Insufficient funds")

        sender.withdraw(amount)
        receiver.deposit(amount)

    def statement(self, account_number):

        account = self.get_account(account_number)

        return account.statement


# Main program

bank = Bank()

bank.create_account("A", 1000)
bank.create_account("B", 200)

bank.transfer("A", "B", 350)

print("Account A:", bank.get_account("A").balance)
print("Account B:", bank.get_account("B").balance)

print("\nAccount A Statement:")

for transaction in bank.statement("A"):
    print(transaction)

print("\nAccount B Statement:")

for transaction in bank.statement("B"):
    print(transaction)
```

```python

from abc import ABC, abstractmethod


class Employee(ABC):

    def __init__(self, name):
        self.name = name

    @abstractmethod
    def calculate_pay(self):
        pass


class SalariedEmployee(Employee):

    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

    def calculate_pay(self):
        return self.salary


class HourlyEmployee(Employee):

    def __init__(self, name, rate, hours):
        super().__init__(name)
        self.rate = rate
        self.hours = hours

    def calculate_pay(self):

        if self.hours > 40:
            regular_pay = 40 * self.rate
            overtime_pay = (self.hours - 40) * self.rate * 1.5
            return regular_pay + overtime_pay

        return self.hours * self.rate


class CommissionEmployee(Employee):

    def __init__(self, name, base, sales, rate):
        super().__init__(name)
        self.base = base
        self.sales = sales
        self.rate = rate

    def calculate_pay(self):
        return self.base + (self.sales * self.rate)


def payroll_summary(employees):

    summary = []

    for employee in employees:

        summary.append({
            "name": employee.name,
            "pay": employee.calculate_pay()
        })

    return summary


# Create employees

employees = [
    SalariedEmployee("Ali", 60000),
    HourlyEmployee("Ahmed", 20, 45),
    CommissionEmployee("Sara", 500, 4000, 0.10)
]


# Generate payroll summary

result = payroll_summary(employees)

for employee in result:
    print(employee)
```

```python

import time
import functools


def ttl_cache(seconds):

    def decorator(func):

        cache = {}
        hits = 0
        misses = 0

        @functools.wraps(func)
        def wrapper(*args, **kwargs):

            nonlocal hits, misses

            try:
                key = (args, tuple(sorted(kwargs.items())))
                hash(key)

            except TypeError:
                raise TypeError("Unhashable arguments are not supported")

            current_time = time.monotonic()

            if key in cache:

                result, saved_time = cache[key]

                if current_time - saved_time < seconds:
                    hits += 1
                    return result

            misses += 1

            result = func(*args, **kwargs)

            cache[key] = (result, current_time)

            return result

        wrapper.cache_info = lambda: {
            "hits": hits,
            "misses": misses
        }

        wrapper.cache_clear = lambda: cache.clear()

        return wrapper

    return decorator


@ttl_cache(seconds=5)
def add(x, y):
    print("Calculating...")
    return x + y


print(add(x=2, y=3))
print(add(y=3, x=2))
print(add(2, 3))

print(add.cache_info())
```

```python

from collections import defaultdict


malformed_count = 0


def read_errors(path):

    global malformed_count

    malformed_count = 0

    with open(path, "r") as file:

        for line_number, line in enumerate(file, start=1):

            parts = line.strip().split(" | ", 2)

            if len(parts) != 3:
                malformed_count += 1
                continue

            timestamp = parts[0]
            level = parts[1]
            message = parts[2]

            if len(timestamp) != 19:
                malformed_count += 1
                continue

            if level not in ["ERROR", "INFO", "WARNING"]:
                malformed_count += 1
                continue

            if level == "ERROR":
                yield line_number, timestamp, message


def summarize_errors(path):

    error_count = defaultdict(int)

    for line_number, timestamp, message in read_errors(path):

        date = timestamp.split(" ")[0]

        error_count[date] += 1

    return dict(error_count)


# Read errors
for error in read_errors("app.log"):
    print(error)


# Summary
summary = summarize_errors("app.log")

print("Errors by date:", summary)
print("Malformed lines:", malformed_count)
```

```python

import csv


inventory = {}

# Read inventory file
with open("inventory.csv", "r", newline="") as file:

    reader = csv.DictReader(file)

    for row in reader:

        sku = row["sku"]

        inventory[sku] = {
            "name": row["name"],
            "quantity": int(row["quantity"])
        }


rejected = []

# Read movements file
with open("movements.csv", "r", newline="") as file:

    reader = csv.DictReader(file)

    for row in reader:

        sku = row["sku"]
        movement_type = row["type"]
        
        try:
            units = int(row["units"])

        except ValueError:
            rejected.append({
                "sku": sku,
                "type": movement_type,
                "units": row["units"],
                "reason": "Invalid units"
            })
            continue

        if sku not in inventory:
            reason = "Unknown SKU"

        elif units <= 0:
            reason = "Units must be positive"

        elif movement_type not in ["IN", "OUT"]:
            reason = "Invalid movement type"

        elif movement_type == "OUT" and inventory[sku]["quantity"] < units:
            reason = "Insufficient stock"

        else:
            reason = None

        if reason is not None:

            rejected.append({
                "sku": sku,
                "type": movement_type,
                "units": row["units"],
                "reason": reason
            })

            continue

        if movement_type == "IN":
            inventory[sku]["quantity"] += units

        elif movement_type == "OUT":
            inventory[sku]["quantity"] -= units


# Write updated inventory
with open("updated_inventory.csv", "w", newline="") as file:

    fieldnames = ["sku", "name", "quantity"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    for sku, details in inventory.items():

        writer.writerow({
            "sku": sku,
            "name": details["name"],
            "quantity": details["quantity"]
        })


# Write rejected movements
with open("rejected_movements.csv", "w", newline="") as file:

    fieldnames = ["sku", "type", "units", "reason"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerows(rejected)


print("Inventory reconciliation completed!")
print("Updated inventory:", inventory)
print("Rejected movements:", rejected)
```

```python

import csv


students = {}
scores = {}
seen_entries = set()
rejected = []


# Read students file
with open("students.csv", "r", newline="") as file:

    reader = csv.DictReader(file)

    for row in reader:

        student_id = row["student_id"]

        students[student_id] = row["name"]
        scores[student_id] = []


# Read scores file
with open("scores.csv", "r", newline="") as file:

    reader = csv.DictReader(file)

    for row in reader:

        student_id = row["student_id"]
        subject = row["subject"]

        try:
            marks = float(row["marks"])

        except ValueError:
            rejected.append({
                "student_id": student_id,
                "subject": subject,
                "reason": "Invalid marks"
            })
            continue

        if student_id not in students:
            reason = "Unknown student ID"

        elif (student_id, subject) in seen_entries:
            reason = "Duplicate subject entry"

        elif marks < 0 or marks > 100:
            reason = "Marks must be between 0 and 100"

        else:
            reason = None

        if reason is not None:

            rejected.append({
                "student_id": student_id,
                "subject": subject,
                "reason": reason
            })
            continue

        seen_entries.add((student_id, subject))
        scores[student_id].append(marks)


# Calculate averages and grades
report = []

for student_id in students:

    if len(scores[student_id]) > 0:

        average = sum(scores[student_id]) / len(scores[student_id])

    else:
        average = 0

    if average >= 85:
        grade = "A"

    elif average >= 70:
        grade = "B"

    elif average >= 55:
        grade = "C"

    elif average >= 40:
        grade = "D"

    else:
        grade = "F"

    report.append({
        "student_id": student_id,
        "name": students[student_id],
        "average": round(average, 2),
        "grade": grade
    })


# Sort by average and student ID
report.sort(key=lambda x: (-x["average"], x["student_id"]))


# Assign ranks
rank = 1

for student in report:
    student["rank"] = rank
    rank += 1


# Export report
with open("report.csv", "w", newline="") as file:

    fieldnames = ["student_id", "name", "average", "grade", "rank"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(report)


print("Gradebook report generated successfully!")

for student in report:
    print(student)


# Show rejected records
print("\nRejected records:")

for row in rejected:
    print(row)
```

```python

class ReversibleRange:

    def __init__(self, start, stop, step=1):

        if step == 0:
            raise ValueError("Step cannot be zero")

        self.start = start
        self.stop = stop
        self.step = step

    def __iter__(self):

        current = self.start

        if self.step > 0:

            while current < self.stop:
                yield current
                current += self.step

        else:

            while current > self.stop:
                yield current
                current += self.step

    def __reversed__(self):

        length = len(self)

        if length == 0:
            return

        current = self.start + (length - 1) * self.step

        for i in range(length):
            yield current
            current -= self.step

    def __len__(self):

        if self.step > 0:

            if self.start >= self.stop:
                return 0

            return (self.stop - self.start + self.step - 1) // self.step

        else:

            if self.start <= self.stop:
                return 0

            return (self.start - self.stop - self.step - 1) // (-self.step)

    def __contains__(self, value):

        if self.step > 0:

            if value < self.start or value >= self.stop:
                return False

        else:

            if value > self.start or value <= self.stop:
                return False

        return (value - self.start) % self.step == 0


# Testing

numbers = ReversibleRange(2, 11, 3)

print("Forward:", list(numbers))
print("Reversed:", list(reversed(numbers)))
print("Length:", len(numbers))
print("5 in numbers:", 5 in numbers)
print("6 in numbers:", 6 in numbers)
```

```python

from datetime import datetime


class BookingScheduler:

    def __init__(self):
        self.bookings = []
        self.booking_ids = set()

    def add_booking(self, room, start, end, booking_id):

        if booking_id in self.booking_ids:
            return "Rejected: Duplicate booking ID"

        if end <= start:
            return "Rejected: Invalid booking time"

        for booking in self.bookings:

            if booking["room"] == room:

                if start < booking["end"] and end > booking["start"]:

                    return {
                        "status": "Rejected",
                        "booking_id": booking_id,
                        "conflict_with": booking["booking_id"]
                    }

        new_booking = {
            "room": room,
            "start": start,
            "end": end,
            "booking_id": booking_id
        }

        self.bookings.append(new_booking)
        self.booking_ids.add(booking_id)

        self.bookings.sort(
            key=lambda x: (x["room"], x["start"])
        )

        return "Booking accepted"

    def show_bookings(self):

        for booking in self.bookings:
            print(booking)


# Create scheduler

scheduler = BookingScheduler()


# Create datetime objects

start1 = datetime(2026, 11, 7, 9, 0)
end1 = datetime(2026, 11, 7, 10, 0)

start2 = datetime(2026, 11, 7, 10, 0)
end2 = datetime(2026, 11, 7, 11, 0)

start3 = datetime(2026, 11, 7, 9, 30)
end3 = datetime(2026, 11, 7, 10, 30)


# Add bookings

print(scheduler.add_booking("A", start1, end1, 101))

print(scheduler.add_booking("A", start2, end2, 102))

print(scheduler.add_booking("A", start3, end3, 103))


# Display accepted bookings

print("\nAccepted Bookings:")

scheduler.show_bookings()
```
