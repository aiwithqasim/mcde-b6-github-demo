Task 01 | Nested Transaction Analyzer

```python
id= {
    1: {
        'customer_name': 'Ali',
        'category': 'Food',
        'amount': 500,
        'status': 'success',
    }
}

id[2]= {
    'customer_name': 'Ali',
            'category': 'Travel',
            'amount': 200,
            'status': 'failed',
        }

id[3]= {
    'customer_name': 'Sara',
    'category': 'Food',
    'amount': 700,
    'status': 'success'
}
```

```python
totals = {}

for transaction in id.values():
    customer = transaction["customer_name"]

    if customer not in totals:
        totals[customer] = 0

    if transaction["status"] == "success":
        totals[customer] += transaction["amount"]

result = []
for customer, total in totals.items():
    result.append({"customer_name": customer, "total": total})

result = sorted(result, key=lambda item: (-item["total"], item["customer_name"]))
```
