Task 06 | Banking Ledger with OOP

```python
class Account:
    next_id = 1

    def __init__(self):
        self._balance = 0
        self._transactions = []

    @property
    def balance(self):
        return self._balance

    def _record(self, action, amount):
        record = (Account.next_id, action, amount, self._balance)
        Account.next_id += 1
        self._transactions.append(record)

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        self._balance += amount
        self._record("deposit", amount)

    def withdraw(self, amount):
        if amount <= 0 or amount > self._balance:
            raise ValueError("Invalid amount or insufficient balance.")
        self._balance -= amount
        self._record("withdraw", amount)

    def statement(self):
        return tuple(self._transactions)


class Bank:
    def __init__(self):
        self._accounts = {}

    def add_account(self, name):
        self._accounts[name] = Account()

    def transfer(self, sender, receiver, amount):
        self._accounts[sender].withdraw(amount)
        self._accounts[receiver].deposit(amount)

    def statement(self, name):
        return self._accounts[name].statement()


bank = Bank()
bank.add_account("Ali")
bank.add_account("Sara")

bank._accounts["Ali"].deposit(100)
bank.transfer("Ali", "Sara", 30)

print(bank._accounts["Ali"].balance)  # 70
print(bank.statement("Ali"))
```
