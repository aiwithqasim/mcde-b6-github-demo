Task 03 | Recursive JSON Path Explorer

```python
def find_paths(data, target):
    paths = []

    def search(value, path):
        if isinstance(value, dict):
            for key, child in value.items():
                search(child, path + "." + str(key) if path else str(key))

        elif isinstance(value, list):
            for index, child in enumerate(value):
                search(child, path + "[" + str(index) + "]")

        else:
            # Keep True separate from 1
            if type(value) == type(target) and value == target:
                paths.append(path)

    search(data, "")
    return paths


data = {"a": [{"x": 5}, 5], "b": {"y": 5}}
print(find_paths(data, 5))
```
