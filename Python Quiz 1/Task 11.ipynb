Task 11 | Multi-File Student Gradebook

```python
import csv

students = [{"student_id": 101, "name": "Student 101"}]

scores = [
    {"student_id": 101, "subject": "Math", "marks": 90},
    {"student_id": 101, "subject": "English", "marks": 80},
    {"student_id": 101, "subject": "Science", "marks": 85},
]

marks_by_id = {student["student_id"]: [] for student in students}
seen = set()

for score in scores:
    student_id = score["student_id"]
    subject = score["subject"]

    if student_id not in marks_by_id or (student_id, subject) in seen:
        continue

    try:
        marks = float(score["marks"])
    except (TypeError, ValueError):
        continue

    if not 0 <= marks <= 100:
        continue

    seen.add((student_id, subject))
    marks_by_id[student_id].append(marks)

report = []

for student in students:
    student_id = student["student_id"]
    marks = marks_by_id[student_id]
    average = sum(marks) / len(marks) if marks else 0

    if average >= 85:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 55:
        grade = "C"
    elif average >= 40:
        grade = "D"
    else:
        grade = "F"

    report.append({
        "student_id": student_id,
        "name": student["name"],
        "average": average,
        "grade": grade,
    })

report.sort(key=lambda student: (-student["average"], student["student_id"]))

for rank, student in enumerate(report, start=1):
    student["rank"] = rank

with open("report.csv", "w", newline="", encoding="utf-8") as file:
    columns = ["student_id", "name", "average", "grade", "rank"]
    writer = csv.DictWriter(file, fieldnames=columns)
    writer.writeheader()
    writer.writerows(report)

print(report)
print("report.csv created")
```
