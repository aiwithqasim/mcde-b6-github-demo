```python
# Q1 Accessing List Elements
numbers = [10, 20, 30, 40, 50, 60, 70]

# Extract the first, middle, and last elements without
#  modifying the original list.
first = numbers[0]
middle = numbers[len(numbers) // 2]
last = numbers[-1]

print(first)   
print(middle)  
print(last)  
```

```python
# Q2 List Slicing
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]

# Create three new lists: first five elements, last four elements, 
# and every second element in reverse order.

first_five = numbers[:5]
last_four = numbers[-4:]
every_second_reverse = numbers[::-2]

print(first_five)
print(last_four)
print(every_second_reverse)
```

```python
# Q3. List Modification

fruits = ["apple", "banana", "cherry", "mango"]

# Replace banana with orange, add grapes at index 2, and remove cherry.

fruits[1] = "orange"
fruits.insert(2, "grapes")
fruits.remove("cherry")

print(fruits)
```

```python
# Q4. Reverse Without reverse()

numbers = [12, 45, 23, 67, 89, 34]

# Reverse using slicing only; preserve the original list.

reversed_numbers = numbers[::-1]

print(reversed_numbers)
print(numbers)
```

```python
# Q5. Second Largest Unique Number

numbers = [45, 12, 89, 34, 89, 67, 45, 23]

# Find the second-largest unique number.

unique_numbers = sorted(set(numbers))
second_largest = unique_numbers[-2]

print(second_largest)
```

```python
# Q6. Rotate a List

numbers = [10, 20, 30, 40, 50, 60]
k = 2

# Rotate the list right by k using slicing.

rotated = numbers[-k:] + numbers[:-k]

print(rotated)

```

```python
# Q7. Remove Duplicates, Preserve Order

numbers = [10, 20, 10, 30, 20, 40, 30, 50]

# Remove duplicates while preserving first-occurrence order

unique_numbers = list(dict.fromkeys(numbers))

print(unique_numbers)
```

```python
# Q8. Flatten a Nested List

data = [[10, 20], [30, 40], [50, 60]]

# Flatten into [10, 20, 30, 40, 50, 60] using list operations.

flattened = data[0] + data[1] + data[2]

print(flattened)

```

```python
# Q9. Extract Data from a Matrix

matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

# Extract main diagonal, secondary diagonal, and middle column; 
# calculate the sum of each.

main_diagonal = [
    matrix[0][0],
    matrix[1][1],
    matrix[2][2]
]

secondary_diagonal = [
    matrix[0][2],
    matrix[1][1],
    matrix[2][0]
]

middle_column = [
    matrix[0][1],
    matrix[1][1],
    matrix[2][1]
]

print("Main diagonal:", main_diagonal)
print("Main diagonal sum:", sum(main_diagonal))

print("Secondary diagonal:", secondary_diagonal)
print("Secondary diagonal sum:", sum(secondary_diagonal))

print("Middle column:", middle_column)
print("Middle column sum:", sum(middle_column))

```

```python
# Q10. List Transformation Challenge

data = [15, 5, 25, 10, 35, 20, 30]

# Build a list of the three largest unique values ascending,
#  then two smallest unique values descending.

unique_sorted = sorted(set(data))

largest_three = unique_sorted[-3:]
smallest_two_desc = unique_sorted[:2][::-1]

result = largest_three + smallest_two_desc

print(result)

```
