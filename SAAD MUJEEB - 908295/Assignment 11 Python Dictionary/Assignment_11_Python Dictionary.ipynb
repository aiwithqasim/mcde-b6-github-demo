```python
# Q31. Dictionary Access

student = {"name": "Ali", "age": 22, "course": "Python", "marks": 85}

# Extract name and marks; retrieve missing city with default Karachi.

name = student["name"]
marks = student["marks"]
city = student.get("city", "Karachi")

print(f"Name: {name}")
print(f"Marks: {marks}")
print(f"City: {city}")
```

```python
# Q33. Dictionary Methods

product = {"id": 101, "name": "Laptop", "price": 120000, "stock": 5}

# Extract keys, values, items; remove stock and retain its value.

keys = product.keys()
values = product.values()
items = product.items()
stock = product.pop("stock")

print(f"Keys: {keys}")
print(f"Values: {values}")
print(f"Items: {items}")
print(f"Removed stock: {stock}")
print(f"Updated product: {product}")
```

```python
# Q35. Student Marks

marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}

# Find highest, lowest, average, and top-scoring student without loops or lambda.

highest = max(marks.values())
lowest = min(marks.values())
average = sum(marks.values()) / len(marks)
top_student = max(marks, key=marks.get)

print(f"Highest Marks: {highest}")
print(f"Lowest Marks: {lowest}")
print(f"Average Marks: {average:.2f}")
print(f"Top Student: {top_student}")
```

```python
# Q36. Nested Employee Records

employees = {
    "E101": {"name": "Ali", "salary": 80000},
    "E102": {"name": "Sara", "salary": 120000},
    "E103": {"name": "Ahmed", "salary": 95000}
}

# Extract Sara salary, increase Ali salary 15%, and calculate total payroll.

sara_salary = employees["E102"]["salary"]

employees["E101"]["salary"] *= 1.15

total_payroll = (
    employees["E101"]["salary"]
    + employees["E102"]["salary"]
    + employees["E103"]["salary"]
)

print(f"Sara's Salary: {sara_salary}")
print(f"Ali's Updated Salary: {employees['E101']['salary']:.0f}")
print(f"Total Payroll: {total_payroll:.0f}")
```

```python
# Q38. Order Calculation

order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}

# Question: Calculate subtotals, grand total, and amount after 10% discount.

laptop_total = order["laptop"]["price"] * order["laptop"]["quantity"]
mouse_total = order["mouse"]["price"] * order["mouse"]["quantity"]
keyboard_total = order["keyboard"]["price"] * order["keyboard"]["quantity"]

grand_total = laptop_total + mouse_total + keyboard_total
discount = grand_total * 0.10
final_amount = grand_total - discount

print(f"Laptop Subtotal: Rs. {laptop_total:,}")
print(f"Mouse Subtotal: Rs. {mouse_total:,}")
print(f"Keyboard Subtotal: Rs. {keyboard_total:,}")
print(f"Grand Total: Rs. {grand_total:,}")
print(f"Discount (10%): Rs. {discount:,.0f}")
print(f"Final Amount: Rs. {final_amount:,.0f}")
```

```python
# Q39. Employee Performance

employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

# Question: Calculate each average, top performer, and highest SQL score.

ali_avg = (
    employees["Ali"]["python"]
    + employees["Ali"]["sql"]
    + employees["Ali"]["aws"]
) / 3

sara_avg = (
    employees["Sara"]["python"]
    + employees["Sara"]["sql"]
    + employees["Sara"]["aws"]
) / 3

ahmed_avg = (
    employees["Ahmed"]["python"]
    + employees["Ahmed"]["sql"]
    + employees["Ahmed"]["aws"]
) / 3

averages = {
    "Ali": ali_avg,
    "Sara": sara_avg,
    "Ahmed": ahmed_avg
}

top_performer = max(averages, key=averages.get)

highest_sql = max(
    employees["Ali"]["sql"],
    employees["Sara"]["sql"],
    employees["Ahmed"]["sql"]
)

print(f"Ali Average: {ali_avg:.2f}")
print(f"Sara Average: {sara_avg:.2f}")
print(f"Ahmed Average: {ahmed_avg:.2f}")
print(f"Top Performer: {top_performer}")
print(f"Highest SQL Score: {highest_sql}")
```

```python
# Q40. Inventory Management

inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

# Question: Calculate total value, most valuable item by price x stock,
#  and update stock after selling two laptops and five mice.

laptop_value = inventory["P101"]["price"] * inventory["P101"]["stock"]
mouse_value = inventory["P102"]["price"] * inventory["P102"]["stock"]
keyboard_value = inventory["P103"]["price"] * inventory["P103"]["stock"]

total_value = laptop_value + mouse_value + keyboard_value

if laptop_value >= mouse_value and laptop_value >= keyboard_value:
    most_valuable = inventory["P101"]["name"]
elif mouse_value >= laptop_value and mouse_value >= keyboard_value:
    most_valuable = inventory["P102"]["name"]
else:
    most_valuable = inventory["P103"]["name"]

inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5

print(f"Total Inventory Value: Rs. {total_value:,}")
print(f"Most Valuable Item: {most_valuable}")
print(f"Laptops Remaining: {inventory['P101']['stock']}")
print(f"Mice Remaining: {inventory['P102']['stock']}")
print(f"Keyboard Stock: {inventory['P103']['stock']}")
```
