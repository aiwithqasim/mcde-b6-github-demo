```python
# Q11. Tuple Indexing

colors = ("red", "green", "blue", "yellow", "purple")

# Extract second, last, and middle three elements.

second = colors[1]
last = colors[-1]
middle_three = colors[1:4]

print("Second element:", second)
print("Last element:", last)
print("Middle three elements:", middle_three)
```

```python
# Q12. Tuple Unpacking

employee = ("Ali", 28, "Data Engineer", 150000)

# Unpack into four variables and display name and salary.

name, age, job, salary = employee

print("Name:", name)
print("Salary:", salary)
```

```python
# Q13. Tuple Concatenation

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# Combine tuples, then create another tuple in reverse order.

combined = tuple1 + tuple2
reversed_tuple = combined[::-1]

print("Combined tuple:", combined)
print("Reversed tuple:", reversed_tuple)
```

```python
# Q14. Count and Locate

numbers = (10, 20, 30, 20, 40, 20, 50)

# Count occurrences of 20 and find the first index.

count_20 = numbers.count(20)
first_index = numbers.index(20)

print("Count of 20:", count_20)
print("First index of 20:", first_index)
```

```python
# Q15. Extended Tuple Unpacking

data = (100, 200, 300, 400, 500, 600)

# Unpack first two and last into variables; put all remaining elements 
# into a list.

first, second, *remaining, last = data

print("First:", first)
print("Second:", second)
print("Remaining:", remaining)
print("Last:", last)
```

```python
# Q16. Modify an Immutable Tuple

numbers = (10, 20, 30, 40, 50)

# Create a new tuple replacing 30 with 300.

new_numbers = numbers[:2] + (300,) + numbers[3:]

print("Original tuple:", numbers)
print("New tuple:", new_numbers)
```

```python
# Q17. Nested Tuple Access

students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)
# Extract Sara second score, Ali average, and Ahmed highest score.
sara_second_score = students[1][1][1]

ali_scores = students[0][1]
ali_average = sum(ali_scores) / len(ali_scores)

ahmed_scores = students[2][1]
ahmed_highest = max(ahmed_scores)

print("Sara's second score:", sara_second_score)
print("Ali's average:", ali_average)
print("Ahmed's highest score:", ahmed_highest)
```

```python
# Q18. Tuple Sorting

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

# Sort by salary descending without loops, comprehensions, or lambda.

from operator import itemgetter

sorted_employees = tuple(
    sorted(employees, key=itemgetter(1), reverse=True)
)

print(sorted_employees)
```

```python
# Q19. Tuple Matrix

matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

# Extract diagonal, calculate its product, and build a tuple of row sums.

diagonal = (
    matrix[0][0],
    matrix[1][1],
    matrix[2][2]
)

diagonal_product = (
    diagonal[0] * diagonal[1] * diagonal[2]
)

row_sums = (
    sum(matrix[0]),
    sum(matrix[1]),
    sum(matrix[2])
)

print("Diagonal:", diagonal)
print("Diagonal product:", diagonal_product)
print("Row sums:", row_sums)
```

```python
# Q20. Inventory Tuple Challenge

records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

# Calculate inventory value, find most expensive product, 
# and build reversed tuple of names.

inventory_value = (
    records[0][1] * records[0][2]
    + records[1][1] * records[1][2]
    + records[2][1] * records[2][2]
)

most_expensive = max(
    records[0],
    records[1],
    records[2],
    key=lambda x: x[1]
)

names = (
    records[0][0],
    records[1][0],
    records[2][0]
)

reversed_names = names[::-1]

print("Total inventory value:", inventory_value)
print("Most expensive product:", most_expensive[0])
print("Reversed product names:", reversed_names)

```
