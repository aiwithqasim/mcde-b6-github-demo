## Q6. Checkout Discount Calculator

Create RegularCustomer, PremiumCustomer, and VIPCustomer classes, each implementing final_price(amount). Apply discounts of 5%, 10%, and 20%, respectively. Write print_receipt(customer, amount) so that it works with any customer type. For a bill of Rs. 10,000, show the final amount for each type. Reject negative bill amounts. Extension: Add GoldCustomer with a 15% discount without modifying print_receipt().

```python
def validate_amount(amount):
    if amount < 0:
        raise ValueError("Bill amount cannot be negative.")

class RegularCustomer:
    def final_price(self, amount):
        validate_amount(amount)
        return amount * (1 - 0.05)

class PremiumCustomer:
    def final_price(self, amount):
        validate_amount(amount)
        return amount * (1 - 0.10)

class VIPCustomer:
    def final_price(self, amount):
        validate_amount(amount)
        return amount * (1 - 0.20)

class GoldCustomer:   
    def final_price(self, amount):
        validate_amount(amount)
        return amount * (1 - 0.15)


def print_receipt(customer, amount):
    final = customer.final_price(amount)
    print(f"{type(customer).__name__:18} Bill: Rs. {amount:,.2f} -> Pay: Rs. {final:,.2f}")


for c in [RegularCustomer(), PremiumCustomer(), VIPCustomer(), GoldCustomer()]:
    print_receipt(c, 10000)

try:
    print_receipt(RegularCustomer(), -500)
except ValueError as e:
    print("Error:", e)
```
