## Q3. Employee Roles

Create a parent class Employee that stores a name and employee ID and provides display_details(). Derive
Developer (programming language) and Manager (team size). Use super() to initialize inherited attributes.
Override display_details() in each child class to include both shared and role-specific information. Create and test one object of each type

```python
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def display_details(self):
        print(f"Name: {self.name}, ID: {self.emp_id}")


class Developer(Employee):
    def __init__(self, name, emp_id, language):
        super().__init__(name, emp_id)
        self.language = language

    def display_details(self):
        super().display_details()
        print(f"Role: Developer, Language: {self.language}")


class Manager(Employee):
    def __init__(self, name, emp_id, team_size):
        super().__init__(name, emp_id)
        self.team_size = team_size

    def display_details(self):
        super().display_details()
        print(f"Role: Manager, Team Size: {self.team_size}")


dev = Developer("Zain", 101, "Python")
mgr = Manager("Hina", 102, 8)
dev.display_details()
print("-" * 30)
mgr.display_details()
```
