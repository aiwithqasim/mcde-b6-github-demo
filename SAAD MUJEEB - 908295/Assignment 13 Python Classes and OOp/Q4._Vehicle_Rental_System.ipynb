## Q4. Vehicle Rental System

Create a base class Vehicle with brand, model, and daily rental rate. Implement calculate_rent(days). Derive Car, which adds a fixed insurance fee of Rs. 1,000 per rental, and Bike, which applies a 10% discount to the base rental amount. Override calculate_rent() and reuse the parent implementation with super(). Reject rental
durations of zero or fewer days. Test both vehicle types for a three-day rental.

```python
class Vehicle:
    def __init__(self, brand, model, daily_rate):
        self.brand = brand
        self.model = model
        self.daily_rate = daily_rate

    def calculate_rent(self, days):
        if days <= 0:
            raise ValueError("Rental days must be greater than zero.")
        return self.daily_rate * days


class Car(Vehicle):
    INSURANCE_FEE = 1000

    def calculate_rent(self, days):
        return super().calculate_rent(days) + self.INSURANCE_FEE


class Bike(Vehicle):
    def calculate_rent(self, days):
        return super().calculate_rent(days) * 0.90


car = Car("Toyota", "Corolla", 5000)
bike = Bike("Honda", "CD70", 800)

print(f"Car rent (3 days): Rs. {car.calculate_rent(3):,.2f}")   
print(f"Bike rent (3 days): Rs. {bike.calculate_rent(3):,.2f}") 

try:
    car.calculate_rent(0)
except ValueError as e:
    print("Error:", e)
```
