## Q7. Shape Calculator

Use ABC and @abstractmethod to define an abstract Shape class with area() and perimeter(). Implement both
methods in Circle and Rectangle. Accept dimensions through constructors and use pi = 3.14159. Test a circle
with radius 5 and a rectangle with width 4 and height 6. Attempt to instantiate Shape directly and explain the
outcome

```python
from abc import ABC, abstractmethod

PI = 3.14159

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return PI * self.radius ** 2

    def perimeter(self):
        return 2 * PI * self.radius


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


c = Circle(5)
r = Rectangle(4, 6)
print(f"Circle    -> Area: {c.area():.2f}, Perimeter: {c.perimeter():.2f}")
print(f"Rectangle -> Area: {r.area():.2f}, Perimeter: {r.perimeter():.2f}")

try:
    s = Shape()
except TypeError as e:
    print("Error:", e)
```
