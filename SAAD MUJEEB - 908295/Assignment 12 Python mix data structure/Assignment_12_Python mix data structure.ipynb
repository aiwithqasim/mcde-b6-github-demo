```python
# Q41. Student Performance System

students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]
# Question: Calculate each average, top student, tuple of names, 
# and set of unique marks.

# Calculate each average
avg1 = sum(students[0]["marks"]) / len(students[0]["marks"])
avg2 = sum(students[1]["marks"]) / len(students[1]["marks"])
avg3 = sum(students[2]["marks"]) / len(students[2]["marks"])

print(f"Ali Average: {avg1:.2f}")
print(f"Sara Average: {avg2:.2f}")
print(f"Ahmed Average: {avg3:.2f}")

# Top student
top_student = max(
    (avg1, students[0]["name"]),
    (avg2, students[1]["name"]),
    (avg3, students[2]["name"])
)
print(f"Top Student: {top_student[1]}")

# Tuple of names
names = (
    students[0]["name"],
    students[1]["name"],
    students[2]["name"]
)
print(f"Tuple of Names: {names}")

# Set of unique marks
unique_marks = (
    set(students[0]["marks"])
    | set(students[1]["marks"])
    | set(students[2]["marks"])
)
print(f"Unique Marks: {unique_marks}")
```

```python
# Q42. Employee Skills

employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

# Question: Find common skills, Ali-only skills, all team skills, 
# and employees meeting Python + SQL requirements.

# Common skills
common_skills = (
    employees["Ali"]
    & employees["Sara"]
    & employees["Ahmed"]
)
print(f"Common Skills: {common_skills}")

# Ali-only skills
ali_only = (
    employees["Ali"]
    - employees["Sara"]
    - employees["Ahmed"]
)
print(f"Ali-only Skills: {ali_only}")

# All team skills
all_skills = (
    employees["Ali"]
    | employees["Sara"]
    | employees["Ahmed"]
)
print(f"All Team Skills: {all_skills}")

# Employees meeting Python + SQL requirements
ali = {"Python", "SQL"}.issubset(employees["Ali"])
sara = {"Python", "SQL"}.issubset(employees["Sara"])
ahmed = {"Python", "SQL"}.issubset(employees["Ahmed"])

print(f"Ali meets requirements: {ali}")
print(f"Sara meets requirements: {sara}")
print(f"Ahmed meets requirements: {ahmed}")
```

```python
# Q43. Sales Data

sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

# Question: Calculate revenue, top-revenue product, name-to-revenue 
# dictionary, and set of prices.

# Calculate revenue
r1 = sales[0][1] * sales[0][2]
r2 = sales[1][1] * sales[1][2]
r3 = sales[2][1] * sales[2][2]

print(f"Laptop Revenue: {r1}")
print(f"Mouse Revenue: {r2}")
print(f"Keyboard Revenue: {r3}")

# Total revenue
total_revenue = r1 + r2 + r3
print(f"Total Revenue: {total_revenue}")

# Top-revenue product
top_product = max(
    (r1, sales[0][0]),
    (r2, sales[1][0]),
    (r3, sales[2][0])
)
print(f"Top-revenue Product: {top_product[1]}")

# Name-to-revenue dictionary
revenue_dict = {
    sales[0][0]: r1,
    sales[1][0]: r2,
    sales[2][0]: r3
}
print(f"Name-to-Revenue: {revenue_dict}")

# Set of prices
prices = {
    sales[0][1],
    sales[1][1],
    sales[2][1]
}
print(f"Set of Prices: {prices}")
```

```python
# Q45. Course Enrollment

courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

# Question: Find students in exactly two, all three, and exactly 
# one course; count unique students.

python_students = courses["Python"]
sql_students = courses["SQL"]
aws_students = courses["AWS"]

# Students in all three courses
all_three = (
    python_students
    & sql_students
    & aws_students
)
print(f"All Three Courses: {all_three}")

# Students in exactly two courses
python_sql = (python_students & sql_students) - aws_students
python_aws = (python_students & aws_students) - sql_students
sql_aws = (sql_students & aws_students) - python_students

exactly_two = python_sql | python_aws | sql_aws
print(f"Exactly Two Courses: {exactly_two}")

# Students in exactly one course
python_only = python_students - sql_students - aws_students
sql_only = sql_students - python_students - aws_students
aws_only = aws_students - python_students - sql_students

exactly_one = python_only | sql_only | aws_only
print(f"Exactly One Course: {exactly_one}")

# Count unique students
unique_students = python_students | sql_students | aws_students
print(f"Unique Students: {len(unique_students)}")
```

```python
# Q46. E-Commerce Transactions

transactions = [
    {"id": 101, "customer": "Ali",
     "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara",
     "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali",
     "items": [("Monitor", 45000, 1)]}
]

# Question: Calculate total revenue, Ali spending, largest transaction, 
# and unique product names.

# Calculate transaction revenues
t1 = (
    transactions[0]["items"][0][1] * transactions[0]["items"][0][2]
    + transactions[0]["items"][1][1] * transactions[0]["items"][1][2]
)

t2 = (
    transactions[1]["items"][0][1] * transactions[1]["items"][0][2]
    + transactions[1]["items"][1][1] * transactions[1]["items"][1][2]
)

t3 = (
    transactions[2]["items"][0][1] * transactions[2]["items"][0][2]
)

# Total revenue
total_revenue = t1 + t2 + t3
print(f"Total Revenue: {total_revenue}")

# Ali spending
ali_spending = t1 + t3
print(f"Ali Spending: {ali_spending}")

# Largest transaction
largest = max(
    (t1, transactions[0]["id"]),
    (t2, transactions[1]["id"]),
    (t3, transactions[2]["id"])
)

print(f"Largest Transaction ID: {largest[1]}")
print(f"Largest Transaction Revenue: {largest[0]}")

# Unique product names
products = (
    {transactions[0]["items"][0][0],
     transactions[0]["items"][1][0]}
    | {transactions[1]["items"][0][0],
       transactions[1]["items"][1][0]}
    | {transactions[2]["items"][0][0]}
)

print(f"Unique Product Names: {products}")
```

```python
# Q47. Pipeline Monitoring

pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

# Question: Find total processed records, failed pipelines, largest 
# pipeline total, and average records per run for each.

# Total records per pipeline
p1 = sum(pipelines["pipeline_1"]["records"])
p2 = sum(pipelines["pipeline_2"]["records"])
p3 = sum(pipelines["pipeline_3"]["records"])

# Total processed records
total = p1 + p2 + p3
print(f"Total Processed Records: {total}")

# Failed pipelines
failed = (
    pipelines["pipeline_1"]["status"] == "failed",
    pipelines["pipeline_2"]["status"] == "failed",
    pipelines["pipeline_3"]["status"] == "failed"
)

failed_pipeline = (
    "pipeline_1" if failed[0]
    else "pipeline_2" if failed[1]
    else "pipeline_3" if failed[2]
    else "None"
)

print(f"Failed Pipeline: {failed_pipeline}")

# Largest pipeline total
largest = max(
    (p1, "pipeline_1"),
    (p2, "pipeline_2"),
    (p3, "pipeline_3")
)

print(f"Largest Pipeline: {largest[1]}")
print(f"Largest Pipeline Total: {largest[0]}")

# Average records per run
avg1 = p1 / len(pipelines["pipeline_1"]["records"])
avg2 = p2 / len(pipelines["pipeline_2"]["records"])
avg3 = p3 / len(pipelines["pipeline_3"]["records"])

print(f"Pipeline 1 Average: {avg1:.2f}")
print(f"Pipeline 2 Average: {avg2:.2f}")
print(f"Pipeline 3 Average: {avg3:.2f}")
```

```python
# Q48. Banking Transactions

accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

# Question: Assuming zero starting balances, calculate final balances, 
# highest balance, total withdrawals, and largest single deposit.

# Final balances
b1 = sum(accounts["A101"]["transactions"])
b2 = sum(accounts["A102"]["transactions"])
b3 = sum(accounts["A103"]["transactions"])

print(f"Ali Final Balance: {b1}")
print(f"Sara Final Balance: {b2}")
print(f"Ahmed Final Balance: {b3}")

# Highest balance
highest = max(
    (b1, "Ali"),
    (b2, "Sara"),
    (b3, "Ahmed")
)

print(f"Highest Balance: {highest[1]} = {highest[0]}")

# Total withdrawals
withdrawals = (
    abs(-2000) + abs(-1000)
    + abs(-3000) + abs(-2000)
    + abs(-1500)
)

print(f"Total Withdrawals: {withdrawals}")

# Largest single deposit
largest_deposit = max(5000, 3000, 10000, 7000, 2500)
print(f"Largest Single Deposit: {largest_deposit}")
```

```python
# Q49. Product Recommendations

users = {
    "Ali": {
        "purchased": {"Laptop", "Mouse"},
        "wishlist": {"Keyboard", "Monitor", "Mouse"}
    },
    "Sara": {
        "purchased": {"Keyboard", "Monitor"},
        "wishlist": {"Laptop", "Mouse"}
    },
    "Ahmed": {
        "purchased": {"Laptop", "Keyboard"},
        "wishlist": {"Monitor", "Headphones"}
    }
}

# Question: Find products bought by at least two users, Ali wishlist
#  minus purchases, shared wishlist products, and all products.

# Products bought by at least two users
ali_bought = users["Ali"]["purchased"]
sara_bought = users["Sara"]["purchased"]
ahmed_bought = users["Ahmed"]["purchased"]

bought_by_two = (
    (ali_bought & sara_bought)
    | (ali_bought & ahmed_bought)
    | (sara_bought & ahmed_bought)
)

print(f"Bought by at Least Two Users: {bought_by_two}")

# Ali wishlist minus purchases
ali_recommendations = (
    users["Ali"]["wishlist"] - users["Ali"]["purchased"]
)

print(f"Ali Wishlist Minus Purchases: {ali_recommendations}")

# Shared wishlist products
shared_wishlist = (
    (users["Ali"]["wishlist"] & users["Sara"]["wishlist"])
    | (users["Ali"]["wishlist"] & users["Ahmed"]["wishlist"])
    | (users["Sara"]["wishlist"] & users["Ahmed"]["wishlist"])
)

print(f"Shared Wishlist Products: {shared_wishlist}")

# All products
all_products = (
    users["Ali"]["purchased"] | users["Ali"]["wishlist"]
    | users["Sara"]["purchased"] | users["Sara"]["wishlist"]
    | users["Ahmed"]["purchased"] | users["Ahmed"]["wishlist"]
)

print(f"All Products: {all_products}")
```

```python
# Q50. Final Company Analysis

company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

# Question:
# 1. Payroll per department.
# 2. Company payroll.
# 3. Highest-paid employee.
# 4. Skills shared by all.
# 5. Engineering-only skills.
# 6. Remaining budget per department.
# 7. Department with higher average salary.
# 8. Name-to-salary dictionary.
# 9. All unique skills.
# 10. Is combined budget sufficient?

eng = company["departments"]["Engineering"]
ana = company["departments"]["Analytics"]

# 1. Payroll per department
eng_payroll = eng["employees"][0][1] + eng["employees"][1][1]
ana_payroll = ana["employees"][0][1] + ana["employees"][1][1]

print(f"Engineering Payroll: {eng_payroll}")
print(f"Analytics Payroll: {ana_payroll}")

# 2. Company payroll
company_payroll = eng_payroll + ana_payroll
print(f"Company Payroll: {company_payroll}")

# 3. Highest-paid employee
highest = max(
    (eng["employees"][0][1], eng["employees"][0][0]),
    (eng["employees"][1][1], eng["employees"][1][0]),
    (ana["employees"][0][1], ana["employees"][0][0]),
    (ana["employees"][1][1], ana["employees"][1][0])
)

print(f"Highest-Paid Employee: {highest[1]}")
print(f"Highest Salary: {highest[0]}")

# 4. Skills shared by all employees
shared_skills = (
    eng["employees"][0][2]
    & eng["employees"][1][2]
    & ana["employees"][0][2]
    & ana["employees"][1][2]
)

print(f"Skills Shared by All: {shared_skills}")

# 5. Engineering-only skills
engineering_skills = (
    eng["employees"][0][2] | eng["employees"][1][2]
)

analytics_skills = (
    ana["employees"][0][2] | ana["employees"][1][2]
)

engineering_only = engineering_skills - analytics_skills
print(f"Engineering-only Skills: {engineering_only}")

# 6. Remaining budget per department
eng_remaining = eng["budget"] - eng_payroll
ana_remaining = ana["budget"] - ana_payroll

print(f"Engineering Remaining Budget: {eng_remaining}")
print(f"Analytics Remaining Budget: {ana_remaining}")

# 7. Department with higher average salary
eng_average = eng_payroll / len(eng["employees"])
ana_average = ana_payroll / len(ana["employees"])

print(f"Engineering Average Salary: {eng_average}")
print(f"Analytics Average Salary: {ana_average}")

higher_average = (
    "Engineering" if eng_average > ana_average
    else "Analytics" if ana_average > eng_average
    else "Equal"
)

print(f"Department with Higher Average Salary: {higher_average}")

# 8. Name-to-salary dictionary
salary_dict = {
    eng["employees"][0][0]: eng["employees"][0][1],
    eng["employees"][1][0]: eng["employees"][1][1],
    ana["employees"][0][0]: ana["employees"][0][1],
    ana["employees"][1][0]: ana["employees"][1][1]
}

print(f"Name-to-Salary Dictionary: {salary_dict}")

# 9. All unique skills
all_skills = (
    eng["employees"][0][2]
    | eng["employees"][1][2]
    | ana["employees"][0][2]
    | ana["employees"][1][2]
)

print(f"All Unique Skills: {all_skills}")

# 10. Is combined budget sufficient?
combined_budget = eng["budget"] + ana["budget"]
sufficient = combined_budget >= company_payroll

print(f"Combined Budget: {combined_budget}")
print(f"Budget Sufficient: {sufficient}")
```
