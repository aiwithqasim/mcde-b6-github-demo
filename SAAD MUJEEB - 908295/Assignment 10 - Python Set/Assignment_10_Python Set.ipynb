```python
# Q21. Unique Values

numbers = [10, 20, 10, 30, 40, 20, 50]

# Question: Convert to a set and count unique elements.

unique_numbers = set(numbers)

print(f"Unique numbers: {unique_numbers}")
print(f"Total unique elements: {len(unique_numbers)}")
```

```python
# Q22. Set Operations

a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

# Find union, intersection, a - b, and symmetric difference.

print(f"Union: {a | b}")
print(f"Intersection: {a & b}")
print(f"a - b: {a - b}")
print(f"Symmetric Difference: {a ^ b}")
```

```python
# Q24. Common Students

python_students = {"Ali", "Sara", "Ahmed", "Zain"}
sql_students = {"Sara", "Ahmed", "Hamza", "Usman"}

# Find students taking both and exactly one course.

both = python_students & sql_students

exactly_one = python_students ^ sql_students

print(f"Students taking both courses: {both}")
print(f"Students taking exactly one course: {exactly_one}")
```

```python
# Q25. Subset and Superset

required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}

# Check required skills are a subset and candidate skills a superset.

print(f"Is subset: {required_skills.issubset(candidate_skills)}")
print(f"Is superset: {candidate_skills.issuperset(required_skills)}")
```

```python
# Q26. Missing Numbers

numbers = {1, 2, 4, 6, 8, 10}

# Find missing integers from 1 through 10 using set operations.

all_numbers = set(range(1, 11))

missing_numbers = all_numbers - numbers

print(f"Missing numbers: {missing_numbers}")
```

```python
# Q27. Three-Set Analysis

python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

# Find students in all three, exactly Python and SQL (not AWS),
#  and at least two.

all_three = python & sql & aws

python_sql_only = (python & sql) - aws

at_least_two = (python & sql) | (python & aws) | (sql & aws)

print(f"Students in all three: {all_three}")
print(f"Python and SQL only: {python_sql_only}")
print(f"Students with at least two skills: {at_least_two}")
```

```python
# Q28. Transaction Analysis

day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

# Find IDs appearing on all days, exactly one day, and at least two days.

all_days = day1 & day2 & day3

at_least_two = (day1 & day2) | (day1 & day3) | (day2 & day3)

all_ids = day1 | day2 | day3

exactly_one = all_ids - at_least_two

print(f"IDs appearing on all days: {all_days}")
print(f"IDs appearing on exactly one day: {exactly_one}")
print(f"IDs appearing on at least two days: {at_least_two}")
```

```python
# Q29. Access Control

required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

# Find missing and extra permissions and test full authorization.

missing = required_permissions - user_permissions

extra = user_permissions - required_permissions

authorized = required_permissions.issubset(user_permissions)

print(f"Missing permissions: {missing}")
print(f"Extra permissions: {extra}")
print(f"Fully authorized: {authorized}")

```

```python
# Q30. Customer Segmentation

website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

# Find customers on exactly one, exactly two, and all three platforms.

all_three = website & mobile & store

exactly_two = (
    ((website & mobile) - store)
    | ((website & store) - mobile)
    | ((mobile & store) - website)
)

all_customers = website | mobile | store

exactly_one = all_customers - (
    (website & mobile) | (website & store) | (mobile & store)
)

print(f"Customers on exactly one platform: {exactly_one}")
print(f"Customers on exactly two platforms: {exactly_two}")
print(f"Customers on all three platforms: {all_three}")
```
