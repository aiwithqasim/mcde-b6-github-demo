```python
ANS=2

numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]

first_five = numbers[:5]
last_four = numbers[-4:]
every_second_reversed = numbers[::-2]

print(first_five)             # [5, 10, 15, 20, 25]
print(last_four)              # [35, 40, 45, 50]
print(every_second_reversed)  # [50, 40, 30, 20, 10]
```

```python
ANS=9

matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90],
]

flat_m = sum(matrix, [])

main_diag = flat_m[::4]          # [10, 50, 90]
secondary_diag = flat_m[2:7:2]   # [30, 50, 70]
middle_col = flat_m[1::3]        # [20, 50, 80]

print("Main diagonal:", main_diag, "sum =", sum(main_diag))            # 150
print("Secondary diagonal:", secondary_diag, "sum =", sum(secondary_diag))  # 150
print("Middle column:", middle_col, "sum =", sum(middle_col))          # 150
```
