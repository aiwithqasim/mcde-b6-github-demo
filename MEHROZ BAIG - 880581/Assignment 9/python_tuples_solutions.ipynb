## Q11. Tuple Indexing
Extract the second, last, and middle three elements.

```python
colors = ("red", "green", "blue", "yellow", "purple")
second = colors[1]
last = colors[-1]
middle_three = colors[1:4]
print("Second:", second)
print("Last:", last)
print("Middle three:", middle_three)
```

## Q13. Tuple Concatenation
Combine the tuples, then create another tuple in reverse order.

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
combined = tuple1 + tuple2
reversed_tuple = combined[::-1]
print("Combined:", combined)
print("Reversed:", reversed_tuple)
```

## Q14. Count and Locate
Count occurrences of 20 and find its first index.

## Q15. Extended Tuple Unpacking
Unpack the first two and last elements; collect the remaining elements in a list.

```python
data = (100, 200, 300, 400, 500, 600)
first, second, *middle, last = data
middle_list = middle
print("First:", first)
print("Second:", second)
print("Middle elements as a list:", middle_list)
print("Last:", last)
```

## Q16. Modify an Immutable Tuple
Create a new tuple replacing 30 with 300.

```python
numbers = (10, 20, 30, 40, 50)
modified = numbers[:2] + (300,) + numbers[3:]
print("Original:", numbers)
print("Modified:", modified)
```

## Q17. Nested Tuple Access
Extract Sara’s second score, Ali’s average, and Ahmed’s highest score.

```python
students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)
sara_second_score = students[1][1][1]
ali_scores = students[0][1]
ali_average = sum(ali_scores) / len(ali_scores)
ahmed_highest_score = max(students[2][1])
print("Sara second score:", sara_second_score)
print("Ali average:", ali_average)
print("Ahmed highest score:", ahmed_highest_score)
```

## Q18. Tuple Sorting
Sort employees by salary descending without loops, comprehensions, or lambda.

```python
employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)
# Sort by the salary field using key extraction and reverse sorting.
# Sorting tuple records directly would sort by name first, so sort salaries
# and reconstruct records by matching each salary to its employee.
salary_order = sorted((employees[0][1], employees[1][1], employees[2][1]), reverse=True)
salary_sorted = (
    employees[0] if employees[0][1] == salary_order[0] else employees[1] if employees[1][1] == salary_order[0] else employees[2],
    employees[0] if employees[0][1] == salary_order[1] else employees[1] if employees[1][1] == salary_order[1] else employees[2],
    employees[0] if employees[0][1] == salary_order[2] else employees[1] if employees[1][1] == salary_order[2] else employees[2],
)
print("Employees by salary descending:", salary_sorted)
```

## Q19. Tuple Matrix
Extract the diagonal, calculate its product, and build a tuple of row sums.

```python
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))
diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])
diagonal_product = diagonal[0] * diagonal[1] * diagonal[2]
row_sums = (sum(matrix[0]), sum(matrix[1]), sum(matrix[2]))
print("Diagonal:", diagonal)
print("Diagonal product:", diagonal_product)
print("Row sums:", row_sums)
```

## Q20. Inventory Tuple Challenge
Calculate total inventory value, find the most expensive product, and reverse the tuple of product names.

```python
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)
laptop_value = records[0][1] * records[0][2]
mouse_value = records[1][1] * records[1][2]
keyboard_value = records[2][1] * records[2][2]
total_inventory_value = laptop_value + mouse_value + keyboard_value
# Compare unit prices without lambda or loops.
most_expensive_product = records[0] if records[0][1] >= records[1][1] and records[0][1] >= records[2][1] else records[1] if records[1][1] >= records[2][1] else records[2]
product_names = (records[0][0], records[1][0], records[2][0])
reversed_names = product_names[::-1]
print("Total inventory value:", total_inventory_value)
print("Most expensive product:", most_expensive_product[0])
print("Product names in reverse:", reversed_names)
```
