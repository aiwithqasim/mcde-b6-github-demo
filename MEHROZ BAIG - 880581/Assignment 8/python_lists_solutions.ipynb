# Python Practice: Lists — Solutions to Q1–Q10



```python
"""
Python Practice: Lists — Solutions to Q1–Q10
"""


def solve_q1():
    # Access the first, middle, and last elements without changing the list.
    numbers = [10, 20, 30, 40, 50, 60, 70]
    first = numbers[0]
    middle = numbers[len(numbers) // 2]
    last = numbers[-1]
    return first, middle, last


def solve_q2():
    # Create the first five, last four, and every second element in reverse order.
    numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
    first_five = numbers[:5]
    last_four = numbers[-4:]
    every_second_reversed = numbers[::-2]
    return first_five, last_four, every_second_reversed


def solve_q3():
    # Replace banana with orange, add grapes at index 2, and remove cherry.
    fruits = ["apple", "banana", "cherry", "mango"]
    fruits[fruits.index("banana")] = "orange"
    fruits.insert(2, "grapes")
    fruits.remove("cherry")
    return fruits


def solve_q4():
    # Reverse using slicing only; preserve the original list.
    numbers = [12, 45, 23, 67, 89, 34]
    reversed_numbers = numbers[::-1]
    return numbers, reversed_numbers


def solve_q5():
    # Find the second-largest unique number.
    numbers = [45, 12, 89, 34, 89, 67, 45, 23]
    unique_numbers = list(set(numbers))
    unique_numbers.sort()
    second_largest = unique_numbers[-2]
    return second_largest


def solve_q6():
    # Rotate right by k using slicing.
    numbers = [10, 20, 30, 40, 50, 60]
    k = 2
    k %= len(numbers)
    rotated = numbers[-k:] + numbers[:-k] if k else numbers[:]
    return rotated


def solve_q7():
    # Remove duplicates while preserving first-occurrence order.
    numbers = [10, 20, 10, 30, 20, 40, 30, 50]
    unique_in_order = list(dict.fromkeys(numbers))
    return unique_in_order


def solve_q8():
    # Flatten one level of nested lists using list operations.
    data = [[10, 20], [30, 40], [50, 60]]
    flattened = data[0] + data[1] + data[2]
    return flattened


def solve_q9():
    # Extract both diagonals and the middle column, then calculate their sums.
    matrix = [
        [10, 20, 30],
        [40, 50, 60],
        [70, 80, 90],
    ]
    main_diagonal = [matrix[0][0], matrix[1][1], matrix[2][2]]
    secondary_diagonal = [matrix[0][2], matrix[1][1], matrix[2][0]]
    middle_column = [matrix[0][1], matrix[1][1], matrix[2][1]]
    sums = {
        "main_diagonal": sum(main_diagonal),
        "secondary_diagonal": sum(secondary_diagonal),
        "middle_column": sum(middle_column),
    }
    return main_diagonal, secondary_diagonal, middle_column, sums


def solve_q10():
    # Three largest unique values ascending, followed by two smallest unique
    # values descending.
    data = [15, 5, 25, 10, 35, 20, 30]
    unique_sorted = sorted(set(data))
    three_largest_ascending = unique_sorted[-3:]
    two_smallest_descending = unique_sorted[:2][::-1]
    result = three_largest_ascending + two_smallest_descending
    return result


def main():
    print("Q1:", solve_q1())
    print("Q2:", solve_q2())
    print("Q3:", solve_q3())
    print("Q4:", solve_q4())
    print("Q5:", solve_q5())
    print("Q6:", solve_q6())
    print("Q7:", solve_q7())
    print("Q8:", solve_q8())
    print("Q9:", solve_q9())
    print("Q10:", solve_q10())


if __name__ == "__main__":
    main()
```
