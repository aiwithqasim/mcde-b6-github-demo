## Q21. Unique Values
Convert the list to a set and count unique elements.

```python
numbers = [10, 20, 10, 30, 40, 20, 50]
unique_numbers = set(numbers)
unique_count = len(unique_numbers)
print("Unique values:", unique_numbers)
print("Number of unique values:", unique_count)
```

## Q22. Set Operations
Find the union, intersection, difference `a - b`, and symmetric difference.

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
union = a | b
intersection = a & b
difference = a - b
symmetric_difference = a ^ b
print("Union:", union)
print("Intersection:", intersection)
print("a - b:", difference)
print("Symmetric difference:", symmetric_difference)
```

## Q23. Add and Remove
Add Snowflake and Databricks; safely remove AWS.

```python
skills = {"Python", "SQL", "AWS"}
skills.update({"Snowflake", "Databricks"})
skills.discard("AWS")
print("Updated skills:", skills)
```

## Q24. Common Students
Find students taking both courses and students taking exactly one course.

```python
python_students = {"Ali", "Sara", "Ahmed", "Zain"}
sql_students = {"Sara", "Ahmed", "Hamza", "Usman"}
both_courses = python_students & sql_students
exactly_one_course = python_students ^ sql_students
print("Both courses:", both_courses)
print("Exactly one course:", exactly_one_course)
```

## Q25. Subset and Superset
Check whether required skills are a subset and candidate skills are a superset.

```python
required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}
is_subset = required_skills.issubset(candidate_skills)
is_superset = candidate_skills.issuperset(required_skills)
print("Required skills are a subset:", is_subset)
print("Candidate skills are a superset:", is_superset)
```

## Q26. Missing Numbers
Find missing integers from 1 through 10 using set operations.

```python
numbers = {1, 2, 4, 6, 8, 10}
all_numbers = set(range(1, 11))
missing_numbers = all_numbers - numbers
print("Missing numbers:", missing_numbers)
```

## Q27. Three-Set Analysis
Find students in all three, exactly Python and SQL (not AWS), and at least two courses.

```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}
all_three = python & sql & aws
python_and_sql_not_aws = (python & sql) - aws
at_least_two = (python & sql) | (python & aws) | (sql & aws)
print("All three:", all_three)
print("Python and SQL but not AWS:", python_and_sql_not_aws)
print("At least two courses:", at_least_two)
```

## Q28. Transaction Analysis
Find IDs appearing on all days, exactly one day, and at least two days.

```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}
all_days = day1 & day2 & day3
exactly_one_day = (day1 | day2 | day3) - ((day1 & day2) | (day1 & day3) | (day2 & day3))
at_least_two_days = (day1 & day2) | (day1 & day3) | (day2 & day3)
print("Appearing on all days:", all_days)
print("Appearing on exactly one day:", exactly_one_day)
print("Appearing on at least two days:", at_least_two_days)
```

## Q29. Access Control
Find missing and extra permissions and test full authorization.

```python
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}
missing_permissions = required_permissions - user_permissions
extra_permissions = user_permissions - required_permissions
fully_authorized = required_permissions.issubset(user_permissions)
print("Missing permissions:", missing_permissions)
print("Extra permissions:", extra_permissions)
print("Fully authorized:", fully_authorized)
```

## Q30. Customer Segmentation
Find customers on exactly one, exactly two, and all three platforms.

```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}
all_three = website & mobile & store
exactly_two = ((website & mobile) | (website & store) | (mobile & store)) - all_three
exactly_one = (website | mobile | store) - ((website & mobile) | (website & store) | (mobile & store))
print("Exactly one platform:", exactly_one)
print("Exactly two platforms:", exactly_two)
print("All three platforms:", all_three)
```
