```python
#Question 1
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

avg1 = sum(students[0]["marks"]) / len(students[0]["marks"])
avg2 = sum(students[1]["marks"]) / len(students[1]["marks"])
avg3 = sum(students[2]["marks"]) / len(students[2]["marks"])

print("Ali Average:", avg1)
print("Sara Average:", avg2)
print("Ahmed Average:", avg3)

highest = max(avg1, avg2, avg3)

if highest == avg1:
    print("Top Student:", students[0]["name"])
elif highest == avg2:
    print("Top Student:", students[1]["name"])
else:
    print("Top Student:", students[2]["name"])

names = (
    students[0]["name"],
    students[1]["name"],
    students[2]["name"]
)
print("Names:", names)

unique_marks = set(students[0]["marks"]) | set(students[1]["marks"]) | set(students[2]["marks"])
print("Unique Marks:", unique_marks)
```

```python
#Question 2
employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

common_skills = employees["Ali"] & employees["Sara"] & employees["Ahmed"]
print("Common Skills:", common_skills)

ali_only = employees["Ali"] - employees["Sara"] - employees["Ahmed"]
print("Ali-only Skills:", ali_only)

all_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]
print("All Team Skills:", all_skills)

required_skills = {"Python", "SQL"}

ali_meets = required_skills.issubset(employees["Ali"])
sara_meets = required_skills.issubset(employees["Sara"])
ahmed_meets = required_skills.issubset(employees["Ahmed"])

print("Ali Meets Requirements:", ali_meets)
print("Sara Meets Requirements:", sara_meets)
print("Ahmed Meets Requirements:", ahmed_meets)
```

```python
#Question 3

sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

revenue1 = sales[0][1] * sales[0][2]
revenue2 = sales[1][1] * sales[1][2]
revenue3 = sales[2][1] * sales[2][2]

total_revenue = revenue1 + revenue2 + revenue3
print("Total Revenue:", total_revenue)

highest_revenue = max(revenue1, revenue2, revenue3)

if highest_revenue == revenue1:
    print("Top Product:", sales[0][0])
elif highest_revenue == revenue2:
    print("Top Product:", sales[1][0])
else:
    print("Top Product:", sales[2][0])

revenue_dict = {
    sales[0][0]: revenue1,
    sales[1][0]: revenue2,
    sales[2][0]: revenue3
}
print("Product Revenue:", revenue_dict)

prices = {sales[0][1], sales[1][1], sales[2][1]}
print("Unique Prices:", prices)
```

```python
#Question 4

customers = {
    "C101": {"name": "Ali", "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara", "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}

total1 = sum(customers["C101"]["orders"])
total2 = sum(customers["C102"]["orders"])
total3 = sum(customers["C103"]["orders"])

spending = {
    "C101": total1,
    "C102": total2,
    "C103": total3
}

print("Spending Per Customer:", spending)

highest = max(total1, total2, total3)

if highest == total1:
    print("Top Customer:", customers["C101"]["name"])
elif highest == total2:
    print("Top Customer:", customers["C102"]["name"])
else:
    print("Top Customer:", customers["C103"]["name"])

totals = (total1, total2, total3)
print("Totals in ID Order:", totals)
```

```python
# Question 5

courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

python = courses["Python"]
sql = courses["SQL"]
aws = courses["AWS"]


all_three = python & sql & aws
print("Students in All Three:", all_three)

python_sql = python & sql
python_aws = python & aws
sql_aws = sql & aws

exactly_two = (python_sql | python_aws | sql_aws) - all_three
print("Students in Exactly Two:", exactly_two)

all_students = python | sql | aws
students_in_two_or_more = python_sql | python_aws | sql_aws

exactly_one = all_students - students_in_two_or_more
print("Students in Exactly One:", exactly_one)

unique_students = python | sql | aws
print("Unique Student Count:", len(unique_students))
```

```python
#Question 6
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

transaction1 = (
    transactions[0]["items"][0][1] * transactions[0]["items"][0][2]
    + transactions[0]["items"][1][1] * transactions[0]["items"][1][2]
)

transaction2 = (
    transactions[1]["items"][0][1] * transactions[1]["items"][0][2]
    + transactions[1]["items"][1][1] * transactions[1]["items"][1][2]
)

transaction3 = (
    transactions[2]["items"][0][1] * transactions[2]["items"][0][2]
)

total_revenue = transaction1 + transaction2 + transaction3
print("Total Revenue:", total_revenue)

ali_spending = transaction1 + transaction3
print("Ali Spending:", ali_spending)

largest = max(transaction1, transaction2, transaction3)
print("Largest Transaction:", largest)

products = {
    transactions[0]["items"][0][0],
    transactions[0]["items"][1][0],
    transactions[1]["items"][0][0],
    transactions[1]["items"][1][0],
    transactions[2]["items"][0][0]
}

print("Unique Products:", products)
```

```python
#Question 7
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

total1 = sum(pipelines["pipeline_1"]["records"])
total2 = sum(pipelines["pipeline_2"]["records"])
total3 = sum(pipelines["pipeline_3"]["records"])

total_processed = total1 + total2 + total3
print("Total Processed Records:", total_processed)

failed_pipelines = set()

if pipelines["pipeline_1"]["status"] == "failed":
    failed_pipelines.add("pipeline_1")

if pipelines["pipeline_2"]["status"] == "failed":
    failed_pipelines.add("pipeline_2")

if pipelines["pipeline_3"]["status"] == "failed":
    failed_pipelines.add("pipeline_3")

print("Failed Pipelines:", failed_pipelines)

largest = max(total1, total2, total3)
print("Largest Pipeline Total:", largest)

average1 = total1 / len(pipelines["pipeline_1"]["records"])
average2 = total2 / len(pipelines["pipeline_2"]["records"])
average3 = total3 / len(pipelines["pipeline_3"]["records"])

print("Pipeline 1 Average:", average1)
print("Pipeline 2 Average:", average2)
print("Pipeline 3 Average:", average3)
```

```python
#Question8
accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

balance1 = sum(accounts["A101"]["transactions"])
balance2 = sum(accounts["A102"]["transactions"])
balance3 = sum(accounts["A103"]["transactions"])

balances = {
    "A101": balance1,
    "A102": balance2,
    "A103": balance3
}

print("Final Balances:", balances)

highest_balance = max(balance1, balance2, balance3)
print("Highest Balance:", highest_balance)

withdrawals = (
    abs(min(accounts["A101"]["transactions"][1], 0))
    + abs(min(accounts["A101"]["transactions"][3], 0))
    + abs(min(accounts["A102"]["transactions"][1], 0))
    + abs(min(accounts["A102"]["transactions"][2], 0))
    + abs(min(accounts["A103"]["transactions"][1], 0))
)

print("Total Withdrawals:", withdrawals)

largest_deposit = max(
    max(accounts["A101"]["transactions"]),
    max(accounts["A102"]["transactions"]),
    max(accounts["A103"]["transactions"])
)

print("Largest Single Deposit:", largest_deposit)
```

```python
#Question 9
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

ali_purchased = users["Ali"]["purchased"]
sara_purchased = users["Sara"]["purchased"]
ahmed_purchased = users["Ahmed"]["purchased"]

bought_by_two_or_more = (
    (ali_purchased & sara_purchased)
    | (ali_purchased & ahmed_purchased)
    | (sara_purchased & ahmed_purchased)
)

print("Bought by at Least Two Users:", bought_by_two_or_more)

ali_recommendations = users["Ali"]["wishlist"] - users["Ali"]["purchased"]
print("Ali's Wishlist Recommendations:", ali_recommendations)

shared_wishlist = (
    (users["Ali"]["wishlist"] & users["Sara"]["wishlist"])
    | (users["Ali"]["wishlist"] & users["Ahmed"]["wishlist"])
    | (users["Sara"]["wishlist"] & users["Ahmed"]["wishlist"])
)

print("Products in at Least Two Wishlists:", shared_wishlist)

all_products = (
    ali_purchased
    | sara_purchased
    | ahmed_purchased
    | users["Ali"]["wishlist"]
    | users["Sara"]["wishlist"]
    | users["Ahmed"]["wishlist"]
)

print("All Products:", all_products)
```

```python
#Question 10
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

engineering = company["departments"]["Engineering"]
analytics = company["departments"]["Analytics"]

# 1. Payroll per department
engineering_payroll = (
    engineering["employees"][0][1]
    + engineering["employees"][1][1]
)

analytics_payroll = (
    analytics["employees"][0][1]
    + analytics["employees"][1][1]
)

print("Engineering Payroll:", engineering_payroll)
print("Analytics Payroll:", analytics_payroll)

# 2. Company payroll
company_payroll = engineering_payroll + analytics_payroll
print("Company Payroll:", company_payroll)

# 3. Highest-paid employee
highest_salary = max(
    engineering["employees"][0][1],
    engineering["employees"][1][1],
    analytics["employees"][0][1],
    analytics["employees"][1][1]
)

if highest_salary == engineering["employees"][0][1]:
    print("Highest Paid:", engineering["employees"][0][0])
elif highest_salary == engineering["employees"][1][1]:
    print("Highest Paid:", engineering["employees"][1][0])
elif highest_salary == analytics["employees"][0][1]:
    print("Highest Paid:", analytics["employees"][0][0])
else:
    print("Highest Paid:", analytics["employees"][1][0])

# 4. Skills shared by all employees
shared_skills = (
    engineering["employees"][0][2]
    & engineering["employees"][1][2]
    & analytics["employees"][0][2]
    & analytics["employees"][1][2]
)

print("Skills Shared by All:", shared_skills)

# 5. Engineering-only skills
engineering_skills = (
    engineering["employees"][0][2]
    | engineering["employees"][1][2]
)

analytics_skills = (
    analytics["employees"][0][2]
    | analytics["employees"][1][2]
)

engineering_only = engineering_skills - analytics_skills
print("Engineering-only Skills:", engineering_only)

# 6. Remaining budget per department
engineering_remaining = engineering["budget"] - engineering_payroll
analytics_remaining = analytics["budget"] - analytics_payroll

print("Engineering Remaining Budget:", engineering_remaining)
print("Analytics Remaining Budget:", analytics_remaining)

# 7. Department with higher average salary
engineering_average = engineering_payroll / len(engineering["employees"])
analytics_average = analytics_payroll / len(analytics["employees"])

if engineering_average > analytics_average:
    print("Higher Average Salary: Engineering")
elif analytics_average > engineering_average:
    print("Higher Average Salary: Analytics")
else:
    print("Both Departments Have Equal Average Salary")

# 8. Name-to-salary dictionary
salary_dict = {
    engineering["employees"][0][0]: engineering["employees"][0][1],
    engineering["employees"][1][0]: engineering["employees"][1][1],
    analytics["employees"][0][0]: analytics["employees"][0][1],
    analytics["employees"][1][0]: analytics["employees"][1][1]
}

print("Name-to-Salary:", salary_dict)

# 9. All unique skills
all_skills = (
    engineering["employees"][0][2]
    | engineering["employees"][1][2]
    | analytics["employees"][0][2]
    | analytics["employees"][1][2]
)

print("All Unique Skills:", all_skills)

# 10. Is combined budget sufficient?
combined_budget = engineering["budget"] + analytics["budget"]

if combined_budget >= company_payroll:
    print("Budget Sufficient: Yes")
else:
    print("Budget Sufficient: No")
```
