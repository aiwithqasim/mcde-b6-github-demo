```python
#Question 7

python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

# Students in all three courses
all = python & sql & aws

python_sql = (python & sql) - aws

two = (python & sql) | (python & aws) | (sql & aws)

print("All three:", all)
print("Python and SQL only:", python_sql)
print("At least two:", two)
```

```python
# Question 8


day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

all_days = day1 & day2 & day3

one_day = (day1 | day2 | day3) - (
    (day1 & day2) | (day1 & day3) | (day2 & day3)
)

two_days = (day1 & day2) | (day1 & day3) | (day2 & day3)

print("All days:", all_days)
print("Exactly one day:", one_day)
print("At least two days:", two_days)
```

```python
# QUestion 9


required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing = required_permissions - user_permissions

extra = user_permissions - required_permissions

authorized = required_permissions.issubset(user_permissions)

print("Missing:", missing)
print("Extra:", extra)
print("Authorized:", authorized)
```

```python
# Question 10


website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website & mobile & store

website_mobile = (website & mobile) - store
website_store = (website & store) - mobile
mobile_store = (mobile & store) - website

exactly_two = website_mobile | website_store | mobile_store

all_customers = website | mobile | store
multiple_platforms = (
    (website & mobile) |
    (website & store) |
    (mobile & store)
)

exactly_one = all_customers - multiple_platforms

print("Exactly one:", exactly_one)
print("Exactly two:", exactly_two)
print("All three:", all_three)
```
