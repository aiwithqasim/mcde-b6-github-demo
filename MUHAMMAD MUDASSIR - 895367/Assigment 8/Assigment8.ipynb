```python
#Question 9


matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

main = [matrix[0][0], matrix[1][1], matrix[2][2]]

secondary = [matrix[0][2], matrix[1][1], matrix[2][0]]

middle = [matrix[0][1], matrix[1][1], matrix[2][1]]

print("Main diagonal:", main)
print("Main sum:", sum(main))

print("Secondary diagonal:", secondary)
print("Secondary sum:", sum(secondary))

print("Middle column:", middle)
print("Middle sum:", sum(middle))
```
