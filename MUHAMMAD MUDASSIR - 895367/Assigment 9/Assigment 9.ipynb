```python
#Question 7


students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)

sara_score = students[1][1][1]

ali_scores = students[0][1]
ali_average = sum(ali_scores) / len(ali_scores)

ahmed_scores = students[2][1]
ahmed_highest = max(ahmed_scores)

print(sara_score)
print(ali_average)
print(ahmed_highest)
```

```python
# Question 9


matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])

product = diagonal[0] * diagonal[1] * diagonal[2]

row1 = sum(matrix[0])
row2 = sum(matrix[1])
row3 = sum(matrix[2])

row_sums = (row1, row2, row3)

print("Diagonal:", diagonal)
print("Product:", product)
print("Row sums:", row_sums)
```

```python

#Question 10
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

laptop = records[0][1] * records[0][2]
mouse = records[1][1] * records[1][2]
keyboard = records[2][1] * records[2][2]

total = laptop + mouse + keyboard

if records[0][1] > records[1][1] and records[0][1] > records[2][1]:
    expensive = records[0]
elif records[1][1] > records[2][1]:
    expensive = records[1]
else:
    expensive = records[2]

names = (records[0][0], records[1][0], records[2][0])
reversed_names = names[::-1]

print("Inventory value:", total)
print("Most expensive product:", expensive)
print("Reversed names:", reversed_names)
```
