```python
ANS=27

python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

all_three = python & sql & aws                                    # {'Ahmed'}
python_sql_not_aws = (python & sql) - aws                         # {'Sara'}
at_least_two = (python & sql) | (python & aws) | (sql & aws)      # {'Sara', 'Ahmed', 'Zain', 'Hamza'}

print(all_three)
print(python_sql_not_aws)
print(at_least_two)
```

```python
ANS=28

day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

all_days = day1 & day2 & day3                                     # {104}
at_least_two = (day1 & day2) | (day1 & day3) | (day2 & day3)      # {103, 104, 106}
exactly_one = (day1 | day2 | day3) - at_least_two                 # {101, 102, 105, 107, 108}

print(all_days)
print(exactly_one)
print(at_least_two)
```

```python
ANS=29

required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing = required_permissions - user_permissions                 # {'execute'}
extra = user_permissions - required_permissions                   # {'delete'}
fully_authorized = required_permissions.issubset(user_permissions)  # False

print(missing)
print(extra)
print(fully_authorized)
```

```python
ANS=30

website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website & mobile & store                              # {'C4'}
at_least_two = (website & mobile) | (website & store) | (mobile & store)
exactly_two = at_least_two - all_three                            # {'C2', 'C3', 'C7'}
exactly_one = (website | mobile | store) - at_least_two           # {'C1', 'C5', 'C6', 'C8'}

print(exactly_one)
print(exactly_two)
print(all_three)
```
