```python
ANS=17

students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)

sara_second = students[1][1][1]

ali_scores = students[0][1]
ali_avg = sum(ali_scores) / len(ali_scores)

ahmed_highest = max(students[2][1])

print(sara_second)    # 87
print(ali_avg)        # 87.66666666666667
print(ahmed_highest)  # 84
```

```python
ANS=18

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

names, salaries = zip(*employees)
sorted_salaries, sorted_names = zip(*sorted(zip(salaries, names), reverse=True))
result = tuple(zip(sorted_names, sorted_salaries))

print(result)  # (('Sara', 120000), ('Ahmed', 95000), ('Ali', 80000))
```

```python
ANS=19

import math

matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

flat = sum(matrix, ())
diagonal = flat[::4]
product = math.prod(diagonal)

r1, r2, r3 = matrix
row_sums = (sum(r1), sum(r2), sum(r3))

print(diagonal)  # (2, 10, 18)
print(product)   # 360
print(row_sums)  # (12, 30, 48)
```

```python
ANS=20

import math

records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

names, prices, qtys = zip(*records)

inventory_value = math.sumprod(prices, qtys)
most_expensive = names[prices.index(max(prices))]
reversed_names = names[::-1]

print(inventory_value)  # 665000
print(most_expensive)   # Laptop
print(reversed_names)   # ('Keyboard', 'Mouse', 'Laptop')
```
