```python
ANS=46

transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]},
]
t1, t2, t3 = transactions
i11, i12 = t1["items"]
i21, i22 = t2["items"]
(i31,) = t3["items"]

txn_total = {
    t1["id"]: i11[1] * i11[2] + i12[1] * i12[2],
    t2["id"]: i21[1] * i21[2] + i22[1] * i22[2],
    t3["id"]: i31[1] * i31[2],
}
total_revenue = sum(txn_total.values())

spend = {}
spend[t1["customer"]] = spend.get(t1["customer"], 0) + txn_total[t1["id"]]
spend[t2["customer"]] = spend.get(t2["customer"], 0) + txn_total[t2["id"]]
spend[t3["customer"]] = spend.get(t3["customer"], 0) + txn_total[t3["id"]]

largest_id = max(txn_total, key=txn_total.get)
product_names = {i11[0], i12[0], i21[0], i22[0], i31[0]}

print(total_revenue)
print(spend["Ali"])
print(largest_id, txn_total[largest_id])
print(product_names)
```

```python
ANS=47

pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)},
}
p1, p2, p3 = pipelines["pipeline_1"], pipelines["pipeline_2"], pipelines["pipeline_3"]

pipe_totals = {
    "pipeline_1": sum(p1["records"]),
    "pipeline_2": sum(p2["records"]),
    "pipeline_3": sum(p3["records"]),
}
total_processed = sum(pipe_totals.values())

failed_flags = {
    "pipeline_1": p1["status"] == "failed",
    "pipeline_2": p2["status"] == "failed",
    "pipeline_3": p3["status"] == "failed",
}
failed = sorted(failed_flags, key=failed_flags.get, reverse=True)[: sum(failed_flags.values())]

largest_pipeline = max(pipe_totals, key=pipe_totals.get)
avg_per_run = {
    "pipeline_1": sum(p1["records"]) / len(p1["records"]),
    "pipeline_2": sum(p2["records"]) / len(p2["records"]),
    "pipeline_3": sum(p3["records"]) / len(p3["records"]),
}

print(total_processed)
print(failed)
print(largest_pipeline, pipe_totals[largest_pipeline])
print(avg_per_run)
```

```python
ANS=48

from bisect import bisect_left

accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]},
}
tx1 = accounts["A101"]["transactions"]
tx2 = accounts["A102"]["transactions"]
tx3 = accounts["A103"]["transactions"]

balances = {"A101": sum(tx1), "A102": sum(tx2), "A103": sum(tx3)}
highest_acc = max(balances, key=balances.get)

all_tx = sorted(tx1 + tx2 + tx3)
neg_count = bisect_left(all_tx, 0)           # kitni values 0 se chhoti hain
total_withdrawals = sum(all_tx[:neg_count])  # sorted mein negatives shuru mein hote hain
largest_deposit = max(all_tx)

print(balances)
print(highest_acc, accounts[highest_acc]["holder"], balances[highest_acc])
print(abs(total_withdrawals))
print(largest_deposit)
```

```python
ANS=49

users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}},
}
A, S, H = users["Ali"], users["Sara"], users["Ahmed"]

bought_2plus = (
    (A["purchased"] & S["purchased"])
    | (A["purchased"] & H["purchased"])
    | (S["purchased"] & H["purchased"])
)
ali_wish_minus_bought = A["wishlist"] - A["purchased"]
shared_wishlist = (
    (A["wishlist"] & S["wishlist"])
    | (A["wishlist"] & H["wishlist"])
    | (S["wishlist"] & H["wishlist"])
)
all_products = (
    A["purchased"] | A["wishlist"]
    | S["purchased"] | S["wishlist"]
    | H["purchased"] | H["wishlist"]
)

print(bought_2plus)
print(ali_wish_minus_bought)
print(shared_wishlist)
print(all_products)
```

```python
ANS=50

company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"}),
            ],
            "budget": 500000,
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"}),
            ],
            "budget": 400000,
        },
    }
}
depts = company["departments"]
eng, ana = depts["Engineering"], depts["Analytics"]
e1, e2 = eng["employees"]
n1, n2 = ana["employees"]

# 1. Payroll per department
payroll = {"Engineering": e1[1] + e2[1], "Analytics": n1[1] + n2[1]}

# 2. Company payroll
company_payroll = sum(payroll.values())

# 8. Name -> salary dictionary
name_to_salary = {e1[0]: e1[1], e2[0]: e2[1], n1[0]: n1[1], n2[0]: n2[1]}

# 3. Highest-paid employee
highest_paid = max(name_to_salary, key=name_to_salary.get)

# 4. Skills shared by all
shared_by_all = e1[2] & e2[2] & n1[2] & n2[2]

# 5. Engineering-only skills
eng_skills = e1[2] | e2[2]
ana_skills = n1[2] | n2[2]
eng_only = eng_skills - ana_skills

# 6. Remaining budget per department
remaining = {
    "Engineering": eng["budget"] - payroll["Engineering"],
    "Analytics": ana["budget"] - payroll["Analytics"],
}

# 7. Department with higher average salary
avg_salary = {
    "Engineering": payroll["Engineering"] / len(eng["employees"]),
    "Analytics": payroll["Analytics"] / len(ana["employees"]),
}
higher_avg_dept = max(avg_salary, key=avg_salary.get)

# 9. All unique skills
all_unique_skills = eng_skills | ana_skills

# 10. Is combined budget sufficient?
combined_budget = eng["budget"] + ana["budget"]
sufficient = combined_budget >= company_payroll

print(payroll)
print(company_payroll)
print(highest_paid, name_to_salary[highest_paid])
print(shared_by_all)
print(eng_only)
print(remaining)
print(higher_avg_dept, avg_salary)
print(name_to_salary)
print(all_unique_skills)
print(sufficient)
```
