# Pillar 1: Encapsulation
### Q1. Student Result Manager

```python
class Student():
    def __init__(self):
        self._marks = str
        self.__grades = str

    def set_marks(self,marks):
        if 0 <= marks <= 100:
            self._marks = marks
        else:
            print("Invalid marks. Marks should be between 0 and 100.")

    def set_grades(self,grades):
        self.__grades = grades

    def get_marks(self):
        return f"Student got {self._marks} marks."

    def get_grades(self):
        if self._marks >= 80:
            return f"Student's garde is A"
        elif self._marks >= 60:
            return f"Student's  garde is B"
        elif self._marks >= 40:
            return f"Student's  garde is C"
        else:
            return f"Student is failed"


std1 = Student()

std1.set_marks(80)
std1.set_grades("A")
print(std1.get_marks())
print(std1.get_grades())
```

# Pillar 2: Inheritance
### Q3. Employee Roles

```python
class Employee():
    def __init__(self,name ,emp_id):
      self.name = name
      self.emp_id = emp_id

    def  display_details(self):
      return f"emplyee name is {self.name} and its id is {self.emp_id}"


class Developer(Employee):
    def __init__(self, name, emp_id, language):
        super().__init__(name, emp_id )
        self.language = language

    def display_details(self):
        return f"{super().display_details()} and its lanuage is {self.language} "


class Manager(Employee):
    def __init__(self, name, emp_id, team_size):
        super().__init__(name, emp_id)
        self.team_size = team_size


    def display_details(self):
        return f"{super().display_details()} and its team_size is {self.team_size} "


dev = Developer("Ali",123,"python")
print(dev.display_details())

mgr = Manager("Ahmed",456,3)
print(mgr.display_details())

```

### Q4. Vehicle Rental System

```python
class Vehicle:
    def __init__(self, brand, model, daily_rate):
        self.brand = brand
        self.model = model
        self.daily_rate = daily_rate

    def calculate_rent(self, days):
        if days <= 0:
            raise ValueError("Rental days must be greater than zero.")
        return self.daily_rate * days

    def show_rent(self, days):
        label = f"{type(self).__name__} | {self.brand} {self.model} | {days} days"
        try:
            print(f"{label}: Rs. {self.calculate_rent(days):,.0f}")
        except ValueError as e:
            print(f"{label}: Error - {e}")


class Car(Vehicle):
    INSURANCE_FEE = 1000

    def calculate_rent(self, days):
        base = super().calculate_rent(days)
        return base + self.INSURANCE_FEE


class Bike(Vehicle):
    DISCOUNT = 0.10

    def calculate_rent(self, days):
        base = super().calculate_rent(days)
        return base * (1 - self.DISCOUNT)



car = Car("Toyota", "Corolla", 5000)
bike = Bike("Honda", "CD70", 1500)

car.show_rent(3)
bike.show_rent(3)
car.show_rent(0)
bike.show_rent(-2)
```

### Q5. Notification Dispatcher

```python
class Notification:
    def send(self, message):
        raise NotImplementedError("Child class must implement send()")

class EmailNotification(Notification):
    def send(self, message):
        print(f"Sending email: {message}")

class SMSNotification(Notification):
    def send(self, message):
        print(f"Sending SMS: {message}")

class PushNotification(Notification):
    def send(self, message):
        print(f"Sending push notification: {message}")

### here is the ploymorphsim in this function
def notify_all(notifications, message):
  for noti in notifications:
    noti.send(message)


channels = [EmailNotification(), SMSNotification(), PushNotification()]
notify_all(channels, "Your order has been shipped!")


class WhatsAppNotification(Notification):
    def send(self, message):
        print(f"💬 WhatsApp message sent: {message}")

print("\n--- After adding WhatsApp ---")
channels.append(WhatsAppNotification())
notify_all(channels, "Your order has been shipped!")
```

### Q6. Checkout Discount Calculator

```python
class RegularCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative.")
        return amount * 0.95


class PremiumCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative.")
        return amount * 0.90


class VIPCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative.")
        return amount * 0.80


def print_receipt(customer, amount):
    print(customer.final_price(amount))


# polymorphism
for c in (RegularCustomer(), PremiumCustomer(), VIPCustomer()):
    print_receipt(c, 10000)

class GoldCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative.")
        return amount * 0.85

print_receipt(GoldCustomer(), 10000)
```

# Pillar 4: Abstraction
### Q7. Shape Calculator

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    PI = 3.14159

    @abstractmethod
    def area(self):
      pass

    @abstractmethod
    def perimeter(self):
      pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return f"The area of Circle is: {self.PI * self.radius ** 2}"

    def perimeter(self):
        return f"The perimeter of circle is: {2 * self.PI * self.radius}"

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return f"The area of rectangle is: {self.length * self.width}"

    def perimeter(self):
        return f"The perimeter of rectangle is: {2 * (self.length + self.width)}"


circle = Circle(5)
print(circle.area())
print(circle.perimeter())

rectangle = Rectangle(4, 6)
print(rectangle.area())
print(rectangle.perimeter())
```

### Q8. Payment Gateway

```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass


class CardPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        print(f"Card , Payment , Rs. {amount}")

    def refund(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        print(f"Card , Refund , Rs. {amount}")


class WalletPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        print(f"Wallet , Payment , Rs. {amount}")

    def refund(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        print(f"Wallet , Refund , Rs. {amount}")


class BankTransferPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        print(f"Bank Transfer , Payment , Rs. {amount}")

    def refund(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        print(f"Bank Transfer , Refund , Rs. {amount}")


def process_transaction(method, amount):
    method.pay(amount)


# Test
for m in (CardPayment(), WalletPayment(), BankTransferPayment()):
    process_transaction(m, 5000)


# Extension
class EasyPaisaPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        print(f"EasyPaisa , Payment , Rs. {amount}")

    def refund(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        print(f"EasyPaisa , Refund , Rs. {amount}")


process_transaction(EasyPaisaPayment(), 5000)
```
