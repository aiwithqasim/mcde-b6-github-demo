## Q2. Secure Bank Account

Create a BankAccount class with a public account holder name, private __balance, and private __pin. Implement
deposit(amount), withdraw(amount, pin), and get_balance(pin). Reject zero or negative deposits, withdrawals
exceeding the available balance, and incorrect PINs. Demonstrate a successful deposit, a successful withdrawal,and two rejected operations. Extension: Explain why double underscores do not provide true security.

```python
class BankAccount:
    def __init__(self, holder_name, initial_balance, pin):
        self.holder_name = holder_name
        self.__balance = initial_balance
        self.__pin = pin

    def deposit(self, amount):
        if amount <= 0:
            print(f"{amount} Deposit rejected: amount must be positive.")
            return
        self.__balance += amount
        print(f"Deposited Rs. {amount}. New balance: Rs. {self.__balance}")

    def withdraw(self, amount, pin):
        if pin != self.__pin:
            print(f"Withdrawal rejected: {pin} incorrect PIN.")
        elif amount <= 0:
            print(f"Withdrawal rejected: {amount} amount must be positive.")
        elif amount > self.__balance:
            print(f"Withdrawal rejected: insufficient balance.")
        else:
            self.__balance -= amount
            print(f"Withdrew Rs. {amount}. Remaining: Rs. {self.__balance}")

    def get_balance(self, pin):
        if pin != self.__pin:
            print("Access denied: incorrect PIN.")
            return None
        return self.__balance


acc = BankAccount("Ahmed", 5000, 1234)
acc.deposit(2000)                 
acc.withdraw(1000, 1234)          
acc.deposit(-500)              
acc.withdraw(100, 9999)          
acc.withdraw(50000, 1234)         
print(f"Balance: Rs. {acc.get_balance(1234)}")

print("Hacked balance:", acc._BankAccount__balance)
print("Hacked PIN:", acc._BankAccount__pin)
```
