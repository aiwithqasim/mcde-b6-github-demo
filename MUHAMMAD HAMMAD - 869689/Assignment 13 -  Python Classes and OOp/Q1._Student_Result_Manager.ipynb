## Q1. Student Result Manager

Create a Student class with a public name, protected _marks, and private __grade attributes. Initialize the
student through a constructor. Implement set_marks(marks), which accepts only values from 0 to 100, along with
get_marks() and get_grade(). Assign grade A for 80 to 100, B for 60 to 79, C for 40 to 59, and F below 40.
Demonstrate both a valid and an invalid marks update. Extension: Attempt direct access to __grade and explain
the result

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self._marks = marks
        self.__grade = "F"
        self.set_marks(marks)

    def set_marks(self, marks):
        if marks >= 0 and marks <= 100:
            self._marks = marks

            if marks >= 80:
                self.__grade = "A"
            elif marks >= 60:
                self.__grade = "B"
            elif marks >= 40:
                self.__grade = "C"
            else:
                self.__grade = "F"
        else:
            print(f"{marks} Invalid Marks!")

    def get_marks(self):
        return self._marks

    def get_grade(self):
        return self.__grade


s1 = Student("Ali", 85)
s2 = Student("Ahmed", 55)

print("Student:", s1.name)
print("Marks:", s1.get_marks())
print("Grade:", s1.get_grade())

print("\nStudent:", s2.name)
print("Marks:", s2.get_marks())
print("Grade:", s2.get_grade())

print("\nUpdating Marks:")
s1.set_marks(95)
print("New Marks:", s1.get_marks())
print("New Grade:", s1.get_grade())

print("\nInvalid Marks:")
s1.set_marks(120)

try:
    print(s1.__grade)
except AttributeError:
    print("Private grade cannot be accessed directly")
```
