## Q8. Payment Gateway



Define an abstract PaymentMethod class with abstract methods pay(amount) and refund(amount). Implement
CardPayment, WalletPayment, and BankTransferPayment. Each method must display the payment method,
transaction type, and amount. Validate that transaction amounts are positive. Write process_transaction(method, amount) to call pay() for any supported payment method. Extension: Add a new payment method without
changing process_transaction().


```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass

    def _validate(self, amount):
        if amount <= 0:
            raise ValueError("Transaction amount must be positive.")


class CardPayment(PaymentMethod):
    def pay(self, amount):
        self._validate(amount)
        print(f"Method: Card | Type: Payment | Amount: Rs. {amount:,.2f}")

    def refund(self, amount):
        self._validate(amount)
        print(f"Method: Card | Type: Refund | Amount: Rs. {amount:,.2f}")


class WalletPayment(PaymentMethod):
    def pay(self, amount):
        self._validate(amount)
        print(f"Method: Wallet | Type: Payment | Amount: Rs. {amount:,.2f}")

    def refund(self, amount):
        self._validate(amount)
        print(f"Method: Wallet | Type: Refund | Amount: Rs. {amount:,.2f}")


class BankTransferPayment(PaymentMethod):
    def pay(self, amount):
        self._validate(amount)
        print(f"Method: Bank Transfer | Type: Payment | Amount: Rs. {amount:,.2f}")

    def refund(self, amount):
        self._validate(amount)
        print(f"Method: Bank Transfer | Type: Refund | Amount: Rs. {amount:,.2f}")


class CryptoPayment(PaymentMethod):  
    def pay(self, amount):
        self._validate(amount)
        print(f"Method: Crypto | Type: Payment | Amount: Rs. {amount:,.2f}")

    def refund(self, amount):
        self._validate(amount)
        print(f"Method: Crypto | Type: Refund | Amount: Rs. {amount:,.2f}")


def process_transaction(method, amount):
    method.pay(amount)


for m in [CardPayment(), WalletPayment(), BankTransferPayment(), CryptoPayment()]:
    process_transaction(m, 2500)

CardPayment().refund(500)

try:
    process_transaction(CardPayment(), -100)
except ValueError as e:
    print("Error:", e)
```
